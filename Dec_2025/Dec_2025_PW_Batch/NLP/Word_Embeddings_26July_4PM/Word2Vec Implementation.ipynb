Load library

```python
from gensim.models import Word2Vec
```

Dataset Creation

```python
sentences = [
    ["the","cat","drinks","milk"],
    ["the","dog","drinks","milk"],
    ["cat","likes","fish"],
    ["dog","likes","bone"],
    ["king","rules","kingdom"],
    ["queen","rules","kingdom"]
]
```

Train

```python
model = Word2Vec(sentences, vector_size=100, window=2, min_count=1, workers=4)
```

Word Vector

```python
model.wv['cat']
```

```python
model.wv.similarity("cat", "dog")
```

```python
model.wv.similarity("king", "queen")
```

```python
model.wv.most_similar("king")
```

```python
model.save('word2vec.model')
```

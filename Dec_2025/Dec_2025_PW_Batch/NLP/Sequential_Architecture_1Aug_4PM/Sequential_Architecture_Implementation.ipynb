Import all the required libraries

Concept of pad_sequences

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

sequences = [
    [10, 25, 8],
    [15, 7, 92, 34, 18],
    [5, 11]
]

pad_sequences(sequences, maxlen=5)
```

Observation: Every sequence has a length of 5. And now, it's easy to pass these sequences to the input layer in the neural network.

Load the data

```python
from tensorflow.keras.datasets import imdb

# keep only the 10,000 most frequently occuring words in the dataset vocabulary.
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words = 10000)
```

Pad sequence

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
max_len = 200

X_train = pad_sequences(X_train, maxlen=max_len)
X_test = pad_sequences(X_test, maxlen=max_len)
```

Build the SimpleRNN model

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
from tensorflow.keras import Input

model = Sequential([
    Input(shape=(200,)),
    Embedding(input_dim = 10000, output_dim = 32),
    SimpleRNN(32),
    Dense(1, activation = 'sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

model.summary()
```

Model Training

```python
history = model.fit(X_train, y_train, epochs = 5, batch_size=128, validation_split=0.2)
```

Model Evaluation

```python
loss, accuracy = model.evaluate(X_test, y_test)
print('Accuracy:', accuracy)
```

Model Prediction

```python
y_pred_prob = model.predict(X_test)
print(y_pred_prob[:5])
```

```python
y_pred_prob = model.predict(X_test)
y_pred = (y_pred_prob > 0.5).astype(int)
print(y_pred)
```

RNN have been widely used in sentiment analysis, chatbot, next word prediction.

# **1. Introduction to Data**

Data is at the heart of every digital system we interact with today. Whether it’s a mobile application suggesting products, a navigation app finding the best route, or a healthcare system predicting risks, all these systems depend on data.

Understanding data, what it is, how it looks, and how it behaves—is the starting point for anyone entering the world of Data Science.

This section introduces the concept of data, explains why it is so valuable, and describes the different forms in which data exists.

## **What is Data?**

Data refers to **any information that is collected, observed, or stored so it can be used later**.
It may represent facts, measurements, records, signals, images, or anything that carries meaning.

Examples from daily life make this easier to understand:

* When you fill out a form with your name and phone number, that becomes data.
* When you take a photo, the image is stored as data on your device.
* When you play a song, the audio file is data.
* When your fitness tracker counts your steps, those numbers are also data.

These examples show that data can be **numerical, textual, visual, or audio-based**.

Data can be extremely simple (like age: 21) or extremely complex (like a 2-minute video). Regardless of its form, data becomes useful only when it is collected, organized, and analyzed.

## **Why Is Data Important in Data Science?**

Data Science is a field that focuses on finding patterns and insights from data. The goal is to answer questions, solve problems, and support decision-making.

Data is important because:

* It tells us **what has happened**
* It helps us understand **why it happened**
* It allows us to predict **what might happen next**

For example:

* **E-commerce platforms** use data to understand customer behavior and show relevant products.
* **Banks** use data to identify suspicious transactions and reduce fraud.
* **Hospitals** analyze patient data to detect health risks early.
* **Transport apps** use location data to calculate travel time.

Data is often called *“the new oil”* because it drives modern businesses and technologies. Even the most advanced machine learning model will fail if the data behind it is poor, incomplete, or incorrect. That makes learning data fundamentals essential for every beginner.

## **Understanding Different Types of Data**

image.png


Real-world data does not come in one single format.
It appears in different shapes, structures, and levels of organization.

Understanding these types is important because each one requires different tools, cleaning methods, and storage approaches.

Below are the four major data categories every beginner should know.

1. Structured Data
2. Unstructured Data
3. Semi-Structured Data.
4. Multi-Model Data

#### **1. Structured Data**

Structured data is the most organized form of data.
It is neatly arranged in **rows and columns**, similar to a spreadsheet.
Each column stores one type of information (such as name, age, or salary), and each row represents one record.

This strict format makes structured data:

* predictable
* easy to navigate
* simple to analyze

#### **Key Characteristics of Structured Data**

* **Clearly defined format:** Everything fits into a table-like structure.
* **Easy to search and filter:** Tools like SQL, Excel, and pandas can quickly find values.
* **Consistent types:** Each column stores a specific type (numbers, dates, categories).
* **High accuracy:** Because data is organized, mistakes are easier to spot.

#### **Common Uses**

Structured data is used in systems where accuracy and consistency are important, such as:

* banking
* billing
* education management
* e-commerce inventory systems

#### **Examples**

* A table of student marks with columns like “Name,” “Math Score,” and “Science Score”
* Employee payroll records
* Online purchase history (customer, product, price, date)
* Hospital patient records

#### **Why Structured Data Is Beginner-Friendly**

Because everything is arranged clearly, it is easy to understand:

* what each column represents
* how many records exist
* how to sort or filter content

This makes structured data the best starting point for new data learners.

### **2. Unstructured Data**

Unstructured data does **not follow a fixed format or layout**.
It cannot be placed into rows and columns directly.

Its content varies widely, from text to images to videos, which makes it harder to analyze using traditional tools.

#### **What Makes It “Unstructured”?**

* There is no predefined arrangement.
* Different parts of the content may not follow patterns.
* The meaning must be interpreted, not extracted directly.

A photo, for example, contains colors and shapes but not labeled columns.
A paragraph of text may express emotion, opinion, or sarcasm things which are not “numeric”.

#### **Common Sources of Unstructured Data**

* social media posts
* WhatsApp messages
* recorded customer calls
* YouTube videos
* photos from mobile phones
* scanned documents
* podcasts

Even within one example, there can be multiple types of information:

A single tweet may include:

* text
* emojis
* hashtags
* images
* GIFs
* a location tag

This flexibility makes unstructured data powerful but also challenging.

#### **Why Unstructured Data Matters**

Today, most of the world’s data, more than **80%** and it is unstructured because people create content naturally:

* writing messages
* posting pictures
* recording videos
* speaking into voice assistants

Businesses use unstructured data to understand human behavior, opinions, and emotions.

### **3. Semi-Structured Data**

Semi-structured data exists **between structured and unstructured data**.
It does not follow a strict table format, but it contains **tags, labels, or markers** that give structure to the content.

#### **What Makes It Semi-Structured?**

* No fixed row-and-column format
* But contains meaningful identifiers
* These identifiers help machines understand the data

For example, an email has:

* *Structured parts:* sender, receiver, timestamp, subject
* *Unstructured part:* the body of the message

This mixture makes the format flexible yet still machine-readable.

### **Examples of Semi-Structured Data**

* **Emails** → structured headers + unstructured body
* **HTML pages** → text wrapped inside tags like `<p>` `<h1>`
* **JSON files** → “key: value” format used in APIs
* **XML files** → used in data sharing systems
* **Sensor logs** → tagged values with timestamps

#### **Why It’s Useful**

Semi-structured data allows:

* flexible content
* easier parsing by machines
* fast data exchange between applications

This format is extremely common in modern apps, websites, and APIs.

### **4. What Is Multimodal Data?**

Multimodal data refers to **data that comes from multiple formats or sensory sources at the same time**.
Instead of relying on only text or only images, multimodal systems combine different data types to build a more complete understanding.

Modern AI systems increasingly use multimodal data because real-world information is rarely limited to a single format.

#### **Examples of Multimodal Data**

#### **1. Social Media Posts**

A single Instagram post may include:

* image (visual data)
* caption (text)
* hashtags (categorical/text)
* location tag (structured data)
* background music (audio)

This is a perfect example of how multiple data types combine to express meaning.

#### **2. YouTube Videos**

A YouTube video includes:

* video frames (image sequence)
* audio track (speech, music)
* title (text)
* description (text)
* comments (text)
* like count, views, and date (structured)

#### **3. Self-Driving Cars**

A self-driving car uses multiple sensors together:

* cameras → images
* LiDAR → 3D spatial/point data
* GPS → location coordinates
* microphones → environmental sounds

Each sensor provides a different perspective needed for safe decision-making.

### **Why Multimodal Data Is Important**

Real-world understanding is never one-dimensional.
A photo without text might be confusing. Text without an image may be incomplete.
Combining different forms of data helps models:

* capture context
* reduce ambiguity
* improve accuracy
* make smarter decisions

Multimodal data is increasingly used in:

* recommendation systems
* content moderation
* virtual assistants
* robotics
* autonomous vehicles
* multimodal AI models (like GPT-4o, Gemini, etc.)

# **2. Data Acquisition**

Data acquisition is the first and most essential stage in any data project. Before we analyze, visualize, or build models, we must gather the data that will support these tasks. This step decides how useful the entire project will be, because every insight depends on the data we collect at the beginning.

In this section, we will understand what data acquisition means, where data typically comes from, how structured and unstructured data are collected, and the challenges involved in this stage.

## **What Does “Data Acquisition” Mean?**

Data acquisition is the **process of collecting data from different sources** so that it can be stored and analyzed later.
This may involve:

* reading information from existing systems
* capturing it through devices
* receiving it from external services
* manually gathering it from people

In simple terms, it is the “data gathering” phase of the data science pipeline.

Just like a building needs the right materials, a data project needs the right data. If the collected data is incomplete, incorrect, or biased, the final results will also be unreliable.

## **Common Ways to Collect Data**

Data can be collected from a variety of sources. Some sources are simple and familiar, while others are advanced and used in large-scale systems. Below are the most common methods, explained in a beginner-friendly way.

image.png

### **1. Files (CSV, Excel, Text Files)**

Files are one of the simplest and most widely used sources of data.
They are easy to store, share, and open.

**CSV (Comma-Separated Values)** and **Excel** files are especially common because they store data in a tabular format. Companies often maintain:

* customer information
* product lists
* sales data
* survey results

Since these files follow a structured format, they are easy to analyze and clean.

### **2. Databases**

A database is a system designed to store large amounts of data in an organized manner.
Think of it like a digital filing cabinet where each drawer and folder has a specific purpose.

Companies use databases because:

* they store data safely
* allow quick searching and updating
* handle millions of records
* support many users at the same time

Examples include systems used in banks, hospitals, e-commerce platforms, and government portals.
Unlike files, databases can grow to huge sizes while maintaining structure and speed.

### **3. APIs (Application Programming Interfaces)**

APIs allow two systems to communicate and share information automatically.

An easy way to understand APIs:

* When you use a weather app, the app itself does not measure the weather.
* Instead, it **calls an API** that provides temperature, humidity, and other details.

APIs help applications receive fresh, real-time data from trusted sources. They are widely used for:

* stock prices
* weather information
* maps and location
* news updates
* social media content

APIs reduce manual work and ensure that data stays up to date.

### **4. Sensors and IoT Devices**

Sensors capture information from the physical world. They convert real-world signals (like temperature, motion, or sound) into data that computers can process.

Common examples include:

* fitness trackers monitoring steps and heart rate
* smart watches measuring temperature and sleep patterns
* cars using sensors to detect speed and distance
* environmental sensors tracking air quality

These devices often collect data continuously, creating a steady stream of information.

### **5. Web Data**

The internet holds vast amounts of information, and many organizations extract data from websites. This process may involve:

* downloading publicly available datasets
* copying data manually
* using tools that scan and extract website information

Web data is useful for tasks like:

* price comparison
* market research
* trend analysis
* product information gathering

While powerful, web data collection requires careful handling to avoid incorrect or outdated information.

### **6. Manual Entry**

In many cases, humans collect and type in data themselves.
This happens in:

* paper or online surveys
* feedback and registration forms
* customer support logs
* medical notes

Manual entry is flexible and often necessary, but human errors such as typing mistakes or incomplete entries are common.

## **Collecting Structured vs Unstructured Data**

Different types of data do not come from the same kind of sources.

Structured data usually comes from organized systems, while unstructured data is generated through more natural, everyday interactions such as taking photos or writing messages. Understanding how these types differ helps us choose the right tools for storing and processing them.

To make this distinction clearer, the table below compares how structured and unstructured data are typically collected.

| Aspect                  | Structured Data                                                                     | Unstructured Data                                             |
| ----------------------- | ----------------------------------------------------------------------------------- | ------------------------------------------------------------- |
| **What it looks like**  | Organized in rows and columns; follows a fixed format                               | No fixed format; free-flowing and varied                      |
| **Common sources**      | Forms with fixed fields, spreadsheets, databases, system logs                       | Chats, social media posts, photos, audio recordings, videos   |
| **How it is collected** | Digital forms, online registrations, business systems, automated logs from software | Cameras, microphones, mobile phones, user uploads, recordings |
| **Ease of analysis**    | Easy to search, filter, and analyze because structure is clear                      | Harder to analyze; needs extra processing to extract meaning  |
| **Examples**            | Customer tables, employee lists, sales records                                      | Images from phones, voice messages, review texts, video clips |



## **Collecting Multimodal Data**

Many modern applications work with more than one type of data at the same time.
This combination is called **multimodal data**.

Depending on the project, multimodal data may include:

* **Text**: chat logs, captions, documents
* **Images**: photographs, scanned documents, medical images
* **Audio**: call recordings, voice messages
* **Video**: surveillance footage, online videos
* **Sensor readings**: temperature, GPS, motion

Collecting multimodal data often requires multiple systems working together. For example, a self-driving car needs video from cameras, distance from sensors, and location from GPS, all collected simultaneously.

## **Data Labeling Basics**

Collecting data alone is not always enough.
For many AI and machine learning tasks, data needs to be **labeled** so the system understands what each example represents.

### **What Is Data Labeling?**

Data labeling means **assigning a meaningful tag or category** to pieces of data.

Examples:

* marking an email as “spam”
* tagging a photo as “cat” or “dog”
* labeling a review as “positive” or “negative”
* identifying whether a transaction is “fraud” or “safe”

Labels act as instructions that tell the model what the data represents.

### **Who Labels the Data?**

Labeling can be done by:

1. **Human annotators** – most accurate, used for complex tasks
2. **Automated tools** – fast, used for repetitive tasks
3. **Semi-automatic systems** – machines suggest labels, humans correct them

Human involvement is common because machines often cannot recognize meaning without guidance.

### **Why Wrong Labeling Affects Model Quality**

Incorrect labels teach a model the wrong patterns.

For instance:

* a dog labeled as a cat
* a genuine review labeled as spam
* a faulty machine reading labeled as normal

Such mistakes cause models to:

* misclassify data
* give unreliable predictions
* behave unfairly or incorrectly

Well-labeled data is one of the strongest foundations for high-quality machine learning systems.

## **Issues in Data Collection**

image.png



Even with the best methods, data collection is rarely perfect.
Several challenges may arise, and understanding them helps improve the quality of collected data.

### **1. Noise**

Noise refers to **unwanted or irrelevant information** added to the data.

Examples:

* blurred images
* background sound in audio recordings
* random characters or spelling mistakes in text

Noise makes it difficult for systems to learn correct patterns.

### **2. Missing Values**

Missing values occur when some data points are not recorded.

Reasons include:

* users skipping questions
* device failure
* sudden connection loss
* incomplete forms

Missing values must be handled carefully during data cleaning.

### **3. Unbalanced Data**

Unbalanced data occurs when certain categories appear much more frequently than others.

Example:

* in a fraud detection dataset, only 1–2% transactions may be fraud.

If not handled properly, models may learn to ignore the rare class entirely.

### **4. Bias During Data Collection**

Bias happens when the collected data does not represent the real world fairly.

Examples:

* collecting opinions only from young people for a nationwide survey
* training a face-recognition system mostly on one skin tone
* collecting customer data only from one city

Biased data leads to biased models, which can produce unfair results.
Avoiding bias is essential for responsible and ethical AI.

# **Data Management**

Once data is collected, the next important step is managing it effectively.

Data management ensures that the data is stored safely, organized properly, and remains easy to access whenever needed. Good data management forms the backbone of every data-driven project. If data is messy, scattered, or insecure, even the most advanced analysis cannot produce reliable results.

This section explains how data should be organized, how different types of storage work, and why privacy and ethics play a crucial role in handling data.

## **What is data management?**

Data management refers to the **process of storing, organizing, protecting, and maintaining data** so it remains useful over time.
It includes everything from choosing where to store the data, to keeping it secure, to ensuring that only the right people can access it.

Good data management helps:

* prevent data loss
* avoid confusion
* reduce errors
* support efficient analysis
* maintain trust and compliance

In simple terms, **data management keeps data safe, structured, and ready to use.**

## **Organizing Data Properly: Files, Folders, and Databases**

image.png


The way data is organized affects how easily it can be accessed or analyzed later.
Here are the basic elements used for storing data.

### **1. Files**

A file stores data in a single unit, such as a CSV, Excel sheet, text document, image, or audio clip. Files are useful for small to medium data and are easy to transfer and share.

### **2. Folders**

Folders help group related files together. For example:

* A “Sales” folder may contain monthly sales reports
* A “Projects” folder may contain documents, images, and notes

Folders help maintain structure and prevent clutter.

### **3. Databases**

Databases are used when data becomes too large or too complex for simple files.
They allow:

* faster searching
* organized storage
* multi-user access
* secure control of information

Companies rely heavily on databases because they keep everything neat, scalable, and reliable.

## **Structured Storage vs Unstructured Storage**

Different data types require different storage systems.
Structured data can fit into traditional databases, while unstructured data needs more flexible storage solutions.

The table below highlights the differences clearly:


| Aspect              | Structured Storage                                   | Unstructured Storage                              |
| ------------------- | ---------------------------------------------------- | ------------------------------------------------- |
| **What it stores**  | Organized tables with rows and columns               | Free-form data like images, text, audio, video    |
| **Storage systems** | Relational databases (MySQL, PostgreSQL, SQL Server) | Cloud storage, data lakes, NoSQL databases        |
| **How data looks**  | Consistent format; easy to index and query           | Irregular format; may contain mixed elements      |
| **Ease of use**     | Very easy to search, filter, update                  | Requires extra processing to extract meaning      |
| **Common examples** | Customer records, transactions, product tables       | Social media posts, scanned documents, recordings |

Structured storage is fast and predictable, while unstructured storage is designed for flexibility and scale.

### **Online vs Offline databases**


| **Point of Difference**  | **Online Database**                                                      | **Offline Database**                                                       |
| ------------------------ | ------------------------------------------------------------------------ | -------------------------------------------------------------------------- |
| **Internet Requirement** | Needs internet or network connection to work.                            | Works without internet; stored on a local machine or device.               |
| **Access**               | Can be accessed by multiple users from different locations.              | Usually accessed by a single user or within one system.                    |
| **Updates**              | Data is updated in real time as users perform actions.                   | Data is updated only when someone manually edits the file or system.       |
| **Data Size Handling**   | Suitable for large, growing, and frequently changing datasets.           | Suitable for smaller datasets with limited growth and simpler needs.       |
| **Use Cases**            | Used in banking systems, e-commerce websites, social media platforms.    | Used in Excel sheets, local inventory tools, offline desktop apps.         |
| **Collaboration**        | Supports team collaboration; many users can read/write at the same time. | Collaboration is limited; sharing usually happens by sending files around. |


### **In-short**

* **Online databases** are best when many people need to use the same data at the same time, often over the internet.

* **Offline databases** are best when data is small, used by few people, and doesn’t need constant online access.

## **Metadata: What It Is and Why It Helps**

Metadata means **“data about data.”**
It gives additional information that helps us understand or organize the main data.

Examples:

* A photo may contain metadata like date, location, and camera type
* A document may show its author and creation time
* A dataset may include column descriptions and units

Metadata helps:

* quickly search files
* identify details without opening the file
* maintain clarity in large systems
* improve data quality during processing

In simple terms, metadata acts like a label or tag that explains the content of the data.

## **Privacy Basics**

When handling data, especially personal information, privacy becomes extremely important.
Data management must ensure that sensitive information is stored securely and used responsibly.

### **Storing Sensitive Data Safely**

Sensitive data includes things like:

* phone numbers
* addresses
* health information
* financial details

This data must be stored with strong protection, such as:

* access control
* encryption
* secure databases
* limited permissions

Keeping such data safe prevents misuse, identity theft, and loss of trust.

### **Why Anonymization Is Needed**

Anonymization means **removing personal details** so that a person cannot be identified.
For example:

* Removing names from survey data
* Hiding phone numbers or exact addresses
* Replacing IDs with random codes

Anonymization protects user identity while still allowing data analysis.
It is often required by law and is a core part of responsible data practices.

## **Ethical Data Use**

Ethical data use ensures that data is handled fairly, transparently, and responsibly.

### **Avoiding Biased or Unfair Datasets**

Data collected from only one group or location may not represent everyone.
This can result in:

* unfair predictions
* inaccurate conclusions
* discrimination in automated systems

Ethical data management aims to reduce such bias by collecting diverse and representative data.

### **Responsible Handling of User Data**

Organizations must:

* collect only necessary data
* use it for the purpose they promised
* avoid sharing it without permission
* delete it when it is no longer needed

Responsible data usage builds trust and ensures that technology benefits everyone.

# **4. Data Cleaning**

After data has been collected and stored, it is rarely ready for direct use. Real-world data almost always contains mistakes, gaps, and inconsistencies. If these problems are ignored, any analysis or model built on top of such data can become misleading or completely wrong.

**Data cleaning** is the process of **finding and correcting problems in the data** so that it becomes accurate, consistent, and reliable. It may not look as “fancy” as building models, but in practice, it is one of the most important steps in data work.

## **Why Raw Data Is Messy**

Raw data becomes messy for many reasons:

* Data is collected from **many different sources** (forms, sensors, websites, manual entry). Each source may follow its own format or rules.
* **Humans make mistakes** while typing, selecting options, or filling forms.
* **Devices and systems fail** - a sensor may stop working, a network may drop, or a web page may not load completely.
* Different people or teams may use **different conventions** for writing the same information (for example, “India”, “IND”, “Bharat”).

Because of these practical reasons, data is often incomplete, duplicated, wrongly formatted, or contains values that do not make sense. Cleaning is the step where we carefully inspect and improve this raw data.

## **Common Problems Found in Data**

image.png

Although every dataset is different, the same types of problems appear again and again. Let us look at them one by one.

### **1. Missing Values**

A missing value means that **no information is available for a particular field**.

Examples:

* A customer skipped the “age” field in a form.
* A temperature sensor failed to record a reading for one hour.
* A transaction record is missing the “payment method”.

Why this is a problem:

* Calculations such as average, sum, or count become incomplete.
* Models may ignore rows with missing values or give errors.
* Patterns can become biased if data is missing more often in some groups than others (for example, younger users not answering income questions).

### **2. Duplicates**

Duplicate records refer to **the same information appearing more than once**.

They often occur when:

* data from multiple systems is merged
* the same form is submitted twice
* logging systems record the same event multiple times

Why this is a problem:

* Counts and totals become inflated (e.g., same person counted twice).
* Patterns may appear stronger than they actually are.
* Storage and processing effort increase without adding new information.

### **3. Wrong Data Types**

Each column or field is supposed to store a particular type of value:
for example, age should be a number, date of birth should be a date, and city should be text.

Problems arise when:

* numerical values are stored as text (e.g., `"25"` instead of `25`)
* dates are stored in mixed formats (`12/05/2024`, `2024-05-12`, `12-May-24`)
* categories are written in many variations (`"Yes"`, `"YES"`, `"Y"`)

Why this is a problem:

* We cannot perform correct calculations, sorting, or comparisons.
* Some tools may treat values as different even when they mean the same thing.
* It becomes hard to join or combine data from multiple sources.

### **4. Inconsistent Formatting**

Inconsistent formatting means **the same type of information is written in different styles** across the dataset.

Examples:

* Country names: `“India”`, `"IND"`, `"Bharat"`
* Phone numbers with and without country code
* Prices with different currency symbols or formats
* Mixed use of lowercase, uppercase, and title case (`"chennai"`, `"Chennai"`)

Why this is a problem:

* Grouping or counting by category becomes inaccurate.
* Searching for a specific value may miss records written in a different style.
* Reports based on such data can be confusing or wrong.

### **5. Outliers**

Outliers are values that are **very far from the rest of the data**.

Examples:

* An age recorded as `250`
* A salary recorded as `1,00,00,00,000` when most are between `20,000` and `1,00,000`
* A product quantity recorded as `1000` instead of `10`

Why this is a problem:

* Outliers can distort averages and trends.
* They might be genuine rare events, or they might simply be data entry errors.
* Without checking, we cannot know whether to trust them or not.

## **How These Problems Are Usually Fixed**

Data cleaning is about making **careful decisions**. There is rarely one single correct method. Instead, we choose an approach that best fits the situation and the goal of the analysis.

Below is a more detailed overview of common strategies.

### **a) Handling Missing Values**

Some common approaches are:

1. **Removing rows or columns**

   * If a row is missing too many important values, it may be better to remove it entirely.
   * If a column is mostly empty and not useful, it may be dropped.
   * This is simple but can reduce the amount of data.

2. **Filling missing values with a fixed value**

   * For example, filling missing city names with `"Unknown"`.
   * Useful when we just need a placeholder and the exact value is not critical.

3. **Filling with summary statistics**

   * For numerical data, we can use the **mean** (average) or **median** (middle value).
   * For categorical data, we can use the **mode** (most common value).
   * This keeps all rows but assumes that missing values are similar to the rest.

4. **Using rules based on context**

   * Example: if “state” is missing but “city” is known, we can infer the state from the city.
   * Requires domain knowledge but leads to more realistic values.

### **b) Dealing with Duplicates**

1. **Detecting exact duplicates**

   * If all column values are the same across two rows, one of them can be safely removed.

2. **Using key fields**

   * Sometimes, we look at key fields like “Customer ID” + “Date” to decide whether two rows represent the same event.

3. **Choosing which record to keep**

   * In some cases, one of the duplicate rows may be more complete or more recent.
   * We keep the best record and remove the rest.

### **c) Correcting Wrong Data Types**

1. **Converting to the correct type**

   * Converting `"25"` (text) to `25` (number).
   * Converting `"2024-11-29"` to an actual date format.

2. **Standardizing categories**

   * Mapping `"YES"`, `"Yes"`, `"Y"` to a single value like `"Yes"`.
   * This makes grouping and counting reliable.

3. **Separating mixed fields**

   * If a field contains both numbers and units (e.g., `"50 kg"`), we may separate it into two fields: `50` and `"kg"`.

### **d) Fixing Inconsistent Formatting**

1. **Standardizing text**

   * Converting all text to a single style: for example, title case (`"Chennai"`) or uppercase.
   * Using a standard list for country names, product names, etc.

2. **Standardizing units and formats**

   * Using the same currency or unit of measurement across the dataset.
   * Converting all date values to one chosen format.

3. **Cleaning extra characters**

   * Removing unwanted spaces, symbols, or repeated punctuation.

### **e) Handling Outliers**

1. **Checking if the outlier is realistic**

   * Sometimes an extreme value is correct (for example, a very high purchase amount for a luxury item).
   * In other cases, it is clearly wrong (age = 250).

2. **Correcting obvious errors**

   * If a decimal point was misplaced or an extra zero was added, we may correct it.

3. **Removing or capping values**

   * For analysis, we may decide to remove outliers that are known to be errors.
   * In some cases, very high or low values are “capped” at a reasonable limit.

The important point is: **outliers should never be removed blindly**. They must be examined carefully.

## **Cleaning Structured vs Unstructured Data**

Cleaning depends not only on the issue, but also on the **type of data**.

Before the table, here is a brief idea:

* **Structured data** (tables) is easier to clean because each value has a clear position and meaning.
* **Unstructured data** (text, images, audio, video) needs more specialized steps to remove noise and extract useful information.

| Aspect                     | Structured Data                                                                                  | Unstructured Data                                                                                                                |
| -------------------------- | ------------------------------------------------------------------------------------------------ | -------------------------------------------------------------------------------------------------------------------------------- |
| **Typical format**         | Rows and columns in tables                                                                       | Free-form text, images, audio, video                                                                                             |
| **Common issues**          | Missing values, duplicates, wrong data types, inconsistent categories                            | Background noise, low-quality images, irrelevant text, mixed languages                                                           |
| **Typical cleaning steps** | Filling or removing missing values, removing duplicates, standardizing formats, correcting types | Text cleaning (removing symbols, stopwords), noise reduction in audio, cropping or enhancing images, removing irrelevant content |
| **Tools usually used**     | Spreadsheet tools, database queries, data frames                                                 | Text processing libraries, image/audio tools, NLP and CV techniques                                                              |
| **Relative difficulty**    | Easier, because structure is clear                                                               | Harder, because meaning is hidden inside complex content                                                                         |

# **5. Exploratory Data Analysis (EDA)**

image.png



Exploratory Data Analysis (EDA) is a critical phase in any data science workflow.

Even after cleaning the data, we cannot assume we fully understand how it behaves. EDA allows us to deeply examine the dataset and uncover insights, patterns, and relationships that would remain hidden if we looked only at raw numbers.

EDA helps us answer essential questions like:

* What does my data look like?
* Is it balanced or skewed?
* Are there unexpected patterns or unusual values?
* What relationships exist between different variables?
* Does the data support the problem I want to solve?

Without performing EDA, building a model is like walking into a dark room — you don’t know what’s inside. EDA “turns the lights on.”

## **What Is EDA and Why It Is Done?**

### **Meaning of EDA**

EDA is a combination of:

* **Summary techniques** (numerical descriptions)
* **Visual techniques** (plots and charts)
* **Logical reasoning** (interpreting what the patterns mean)

It helps us explore the *shape*, *behavior*, and *relationships* within the data.

### **Why EDA is essential**

EDA is done because:

1. **Raw numbers do not reveal everything.**
   Patterns become visible only when we summarize or visualize the data.

2. **EDA tells us whether the dataset is trustworthy.**
   Unexpected spikes or gaps may signal collection issues.

3. **EDA helps us discover valuable patterns.**
   These patterns often guide business decisions or feature engineering steps.

4. **EDA helps identify relationships between variables.**
   For example, higher age → more spending, or higher traffic → lower sales.

5. **EDA suggests which type of model might work best.**
   If relationships appear linear, then linear models may work; if patterns are irregular, decision trees might fit better.

6. **EDA helps us catch errors early.**
   It's better to detect issues now than after building a model.

**In short: EDA builds understanding, direction, and confidence.**

## **What We Check During EDA**


Each component of EDA reveals something different. Let’s break each one down with more depth and clarity.

### **1. Dataset Shape**

When we look at a dataset, the first question is:
**How big is it, and what structure does it have?**

Dataset shape tells us:

* **Number of rows** → how many observations/data points
* **Number of columns** → how many features/attributes

#### **Why does this matter?**

* A dataset with *too few rows* may not be representative.
* A dataset with *too many columns* may contain unnecessary or irrelevant features.
* A very large dataset may require efficient tools (like chunking, sampling).

#### **Example:**

A customer dataset with:

* 30,000 rows
* 18 columns

This tells us:

* There is enough data to find trends.
* Many types of customer information are available to study.

### **2. Column Types**

Identifying column types is crucial because different data types need different treatment.

Common types:

| Type        | Example                    | How we treat it           |
| ----------- | -------------------------- | ------------------------- |
| Numerical   | Age, salary, marks         | Histograms, mean, median  |
| Categorical | City, gender, product type | Bar charts, counts        |
| Date/Time   | Purchase date, login time  | Line charts, trends       |
| Text        | Reviews, comments          | Word frequency, sentiment |
| Boolean     | Yes/No fields              | Bar charts, counts        |

#### **Why column types matter**

* Numerical values are used in calculations.
* Categorical values need encoding before modeling.
* Dates reveal time trends.
* Text and images require special preprocessing.

If column types are incorrect (e.g., a number stored as text), EDA results will be misleading.

### **3. Summary Statistics**

These are high-level numerical summaries that give immediate insights.

For numerical data, we look at:

* **Mean** → average
* **Median** → central value
* **Minimum & Maximum** → extreme values
* **Standard deviation** → how spread out values are
* **Quartiles/Percentiles** → distribution shape

#### **Why summary statistics matter**

* They highlight unusual values (e.g., max income = ₹10,00,000 but most earn ₹30,000).
* They reveal whether the data is skewed or balanced.
* They show how stable or variable the values are.
* They help us detect input errors (negative age, unrealistic heights, etc.).

Summary stats create the *foundation* for deeper exploratory work.

### **4. Patterns & Relationships**

EDA is not just about individual features, it also examines how features interact.

We look for:

* **Trends over time** (e.g., sales increasing monthly)
* **Group differences** (e.g., one city buying more than another)
* **Correlations** (e.g., price ↑ → demand ↓)
* **Clusters** (groups with similar behavior)
* **Unexpected spikes** (seasonal behavior or errors)

### **Real example:**

If younger customers buy more during weekends, this is a behavioral pattern that can guide marketing decisions.

Patterns help us understand:

* **“What is happening?”**
* **“Why is it happening?”**
* **“What should we do next?”**

## **Types of EDA**

Exploratory Data Analysis can be performed in different ways depending on how many variables we study at once.

Each type of analysis answers different questions about the data. Understanding these types clearly helps students know when and why to use each method, especially in future practical work.



image.png

### **1. Univariate Analysis (Studying One Variable)**

Univariate analysis focuses on understanding **one variable at a time**.
It is the simplest form of analysis, but it provides essential insights about a dataset.

#### **What does univariate analysis help us understand?**

1. **Distribution of values**
   Whether the values are centered, spread out, or skewed to one side.

2. **Most frequent and least frequent values**
   Helps in identifying dominant categories or common patterns.

3. **Presence of unusual or extreme values**
   Outliers become visible when we study a single variable in isolation.

4. **Shape of the data**
   Whether the data looks normal, uniform, or irregular.

5. **Balance within categories**
   For groups like gender, city, or product type.

#### **Why is this important?**

* It builds understanding of each column individually.
* It tells us whether a feature is useful or needs modification.
* It guides cleaning decisions like outlier handling or transformations.

#### **Examples of univariate questions:**

* What is the most common age group?
* How many orders per city?
* Are product ratings mostly positive or negative?
* What price range appears most often?

Univariate analysis forms the foundation before comparing features with each other.

### **2. Bivariate Analysis (Studying Two Variables Together)**

Bivariate analysis explores **the relationship between two variables** at the same time.
This is where deeper insights usually emerge, helping us understand how one variable might influence another.

#### **What does bivariate analysis help us discover?**

1. **Correlation**
   Whether two variables move together (e.g., income ↑ → spending ↑).

2. **Group comparisons**
   How different categories perform across another feature
   (e.g., average purchase amount by city).

3. **Cause-and-effect tendencies**
   Though not perfect proof, bivariate patterns often reveal influence.

4. **Patterns and clusters**
   Groups of data points that behave similarly.

5. **Anomalies and unexpected behavior**
   Points that do not follow the general pattern.

#### **Why is this important?**

* Helps select relevant features for modeling.
* Builds understanding of business behavior.
* Shows which variables have meaningful connections.
* Reveals hidden patterns not visible in univariate analysis.

#### **Examples of bivariate questions:**

* Does age influence online spending?
* Do people from certain cities buy more frequently?
* Does marketing spend correlate with sales?
* How does price affect number of units sold?

Understanding 2-variable relationships is key for feature engineering and predictive modeling.

# **Feature Engineering**

Feature engineering is the step where we **shape the data so that a model can actually learn from it**.

We do not just “feed the dataset” to a model and hope for the best. Instead, we carefully think about:

* what information each column carries
* whether that information is in a useful form
* what new information we can create from existing columns

Feature engineering is often where **human understanding of the problem** is combined with data. That is why it is such an important skill.


## **What Are Features?**

In simple words, a **feature** is a property, characteristic, or measurement that describes an object, person, event, or record in the dataset.

In a table:

* each **row** = one data record (one customer, one transaction, one image, etc.)
* each **column** = one feature

#### **Examples**

a) For a customer data table, features might be:

  * age
  * gender
  * city
  * total amount spent
  * number of orders

b) For a house price dataset, features might be:

* area in square feet
* number of bedrooms
* location
* age of the building
* distance to main road

c) For an image, the raw features are:

* pixel values (each pixel has an intensity or color)

d) For text, the raw features are:

* characters or words in the sentence

When we talk about “engineering features”, we mean:

> choosing, cleaning, transforming, and creating these columns so they represent the problem in the best possible way.

## **Why Good Features Matter**

A machine learning model is like a student preparing for an exam.

* If the student receives **clear notes with key concepts**, they will perform well.
* If the notes are **messy, incomplete, or misleading**, even a smart student will struggle.

The model is the student.
The **features are the notes**.

#### **What good features do**

Good features:

* highlight meaningful patterns
* separate different classes clearly (e.g., “fraud” vs “non-fraud”)
* reduce noise and confusion
* help the model learn faster and more accurately

#### **What bad features cause**

Poorly engineered features can:

* hide the true pattern (e.g., mixing different units, mis-encoded categories)
* confuse the model (e.g., random IDs as inputs)
* add noise (irrelevant or duplicate information)
* make the model overfit (memorize useless details instead of learning general rules)

In real projects, **improving features often increases performance more than changing the model type**.

## **Basic Feature Engineering Tasks**

Now let’s look at core tasks you will apply frequently.



image.png

### **1. Cleaning Categories**

Categorical features are columns with fixed groups or labels, such as:

* city
* product category
* payment method
* marital status

These values often come from human input or multiple systems, so they are full of small differences that matter to computers.

#### **Common problems**

* Same value written in different ways

  * `"Bangalore"`, `"Bengaluru"`, `"bangalore"`

* Typos

  * `"Cash"`, `"Cahs"`

* Extra spaces or symbols

  * `" Delhi"`, `"Delhi "`

* Mixed language or style

  * `"Yes"`, `"YES"`, `"Y"`

To a computer, these are all **different categories**, which breaks grouping and statistics.

#### **Typical cleaning steps**

* convert all text to a consistent case (e.g., all lowercase)

* trim spaces at the start and end

* map equivalent labels to a single standard form

  * `"Bangalore"` and `"Bengaluru"` → `"Bengaluru"`

* remove strange symbols or emojis (if not meaningful)

#### **Simple example**

**Raw data:**

| City      |
| --------- |
| Bangalore |
| Bengaluru |
| bangalore |
| Bengaluru |
| BENGALURU |

**After cleaning and standardization:**

| City      |
| --------- |
| Bengaluru |
| Bengaluru |
| Bengaluru |
| Bengaluru |
| Bengaluru |

Now the model sees one clear category instead of many messy variations.

### **2. Handling Numerical Values**

Numerical features are columns that contain numbers, such as:

* age
* salary
* amount spent
* number of items

Even though they look simple, they have their own issues.

#### **Common issues**

* **Missing values** → some rows have no value
* **Outliers** → extreme values (age 250, salary 10 crore)
* **Different scales** → “amount spent” ranges from 0–10,000, while “age” ranges from 18–70
* **Wrong unit or format** → “height” in cm in some rows and inches in others

#### **How we handle them**

* deal with missing values (drop, fill, or estimate)
* check for unrealistic values and correct/remove them
* standardize or scale values if needed (so that columns with large numbers don’t dominate certain models)
* ensure units are consistent (all in meters, all in rupees, etc.)

The goal is to make each numerical column **trustworthy and comparable**.

### **3. Creating New Meaningful Features**

Sometimes, the most powerful features do not exist in the raw data. We create them using domain knowledge and logical combinations.

This is where human understanding adds real value.

#### **Types of new features**

1. **Combination features**

   * Total amount spent = `quantity × price`
   * BMI = `weight / (height²)`

2. **Ratio or rate features**

   * Average order value = `total_spent / number_of_orders`
   * Click-through rate = `clicks / impressions`

3. **Time-based features**

   * Day of week (Monday–Sunday)
   * Hour of day (0–23)
   * Weekend vs weekday
   * Time since last purchase

4. **Count features**

   * Number of logins in last 30 days
   * Number of support tickets raised by a user

5. **Flags (Yes/No indicators)**

   * “High value customer?” → Yes if total_spent > some threshold
   * “Late payment?” → Yes if due_date < payment_date

#### **Why this matters**

Creating new features often:

* reveals patterns that were hidden
* simplifies the model’s job
* makes the outcome more explainable to humans

Example: instead of giving separate “height” and “weight” columns, giving BMI may directly relate better to a health risk label.

### **4. Binning and Grouping**

Binning means turning continuous numerical values into **ranges** or **categories**.

#### **Example: Age**

Instead of using exact age:

* 17, 18, 19, 23, 42, 59…

We can create bins:

* 0–18  → “Teen”
* 19–35 → “Young Adult”
* 36–50 → “Middle Age”
* 51+   → “Senior”

#### **Why binning is useful**

* makes interpretation easier (“most bookings are from young adults”)
* reduces sensitivity to small variations (age 21 vs 22 is often not important)
* can reduce noise and smooth out irregularities
* sometimes helps models that struggle with non-linear relationships

#### **Caution**

Too much binning can also throw away useful detail.
So we use it **when simplification helps** and does not destroy important information.

## **Feature Encoding**

Most machine learning models work only with **numbers**.
They cannot directly work with text labels like `"Delhi"` or `"Gold"`.

Feature encoding converts **categorical features into numerical representations**.

We focus on two basic encoding methods: **Label Encoding** and **One-Hot Encoding**.

### **1. Label Encoding**

Label Encoding gives each category a unique number.

#### **Example**

**Raw categories:**

| Size   |
| ------ |
| Small  |
| Medium |
| Large  |

**After label encoding:**

| Size   | Size_Label |
| ------ | ---------- |
| Small  | 0          |
| Medium | 1          |
| Large  | 2          |

Now the model sees 0, 1, 2 instead of words.

#### **When label encoding makes sense**

* when categories have a **natural order**

  * Small < Medium < Large
  * Beginner < Intermediate < Advanced

In such cases, the numerical order reflects the real-world order.

#### **Risk of misuse**

If we apply label encoding to **unordered categories**, like:

* `"Red"`, `"Blue"`, `"Green"`

and encode them as 0, 1, 2, the model might mistakenly assume that:

* 2 (“Green”) > 1 (“Blue”) > 0 (“Red”)

This introduces a fake ordering that does not exist in reality.

### **2. One-Hot Encoding**

One-Hot Encoding avoids the “fake order” problem by creating a new column for each category.

Each row gets:

* 1 in the column corresponding to its category
* 0 in the other category columns

#### **Example**

**Raw categories (City):**

| City    |
| ------- |
| Delhi   |
| Mumbai  |
| Chennai |

**After one-hot encoding:**

| City    | City_Delhi | City_Mumbai | City_Chennai |
| ------- | ---------- | ----------- | ------------ |
| Delhi   | 1          | 0           | 0            |
| Mumbai  | 0          | 1           | 0            |
| Chennai | 0          | 0           | 1            |

The model now sees **presence (1)** or **absence (0)**, without any order between cities.

#### **When one-hot encoding is preferred**

* when categories **do not have a natural order**
* when we want to avoid wrong assumptions about distance between categories
* in many linear models and neural networks

#### **Trade-off**

* One-hot encoding is very safe and clear.
* But if there are **too many categories**, it can create a very wide dataset (many columns).

## **Feature Store**


image.png



A **feature store** is a place where a company **saves useful features** so they can be reused for different machine learning models.

Think of it as a well-organized **library of features**.

Instead of creating the same features again and again, the feature store keeps them ready for future use.


#### **Why a Feature Store Is Helpful**

* Saves time — features don’t need to be rebuilt from scratch.
* Keeps features **consistent** across different projects.
* Helps teams reuse good features easily.

#### **A Simple Example**

Imagine you work in a company’s data team.
You often create features like:

* How many times a customer visited the website in the last month
* Total amount a customer spent
* Whether a customer is an active user or not

You use these features to build a model.
Now another team also wants to build a different model, and they need the **same features**.

Without a feature store:

* Each team repeats the work
* The same feature may be calculated differently
* Results can become inconsistent

With a feature store:

* These commonly used features are stored in one place
* Anyone building a new model can **take the ready-made features**
* No repetition, no confusion, and faster model building.

#### **Where It Fits in the ML Workflow**

A feature store helps **after features are created** and supports:

* Training machine learning models
* Reusing the same features in future models
* Keeping all features organized and consistent

#### **In Simple Words**

A feature store helps a team work smarter by saving:

* **Time**
* **Effort**
* **Repeated work**
* **Confusion**

It makes feature engineering more organized and efficient.

## **Feature Engineering for Unstructured Data**


image.png



So far, most examples involved structured/tabular data.
But we also work with **text, images, audio, etc.** These are called *unstructured data* and need special feature engineering.

### **1. Text Feature Engineering**

Raw text is messy and not directly usable by models. Before we talk about tokenization and embeddings, we can create simple, classical text features.

#### **Common simple text features**

* **Length of text**

  * number of characters
  * number of words

* **Count of specific words**

  * how many times “refund”, “angry”, or “delay” appears

* **Presence of patterns**

  * whether text contains a question mark
  * whether it contains an email ID or phone number

* **Bag-of-words counts (conceptually)**

  * how many times each word appears in a document
  * later converted to a numerical vector

These features help basic models get some sense of the structure and content of text.

### **2. Image Feature Engineering**

For images, feature engineering begins with understanding their structure.

An image is:

* a grid of pixels (height × width)
* each pixel has intensity (in grayscale) or color values (in RGB)

#### **Basic feature-related concepts**

* **Image dimensions**

  * e.g., 224 × 224 pixels
  * many models require all images to be resized to the same dimensions

* **Color channels**

  * grayscale → 1 channel
  * color (RGB) → 3 channels

* **Basic features we might derive** (conceptual only)

  * average brightness
  * color dominance (more red/blue/green)
  * simple texture or edge measures

In deep learning, later layers automatically learn complex features, but understanding these basic ideas helps students see images as structured data, not magic.

## **GenAI-Specific Preprocessing (Tokenization & Embeddings)**

Generative AI models (for text, images, or multimodal tasks) rely heavily on **special kinds of feature engineering**.

They cannot work directly with raw text or pixels.
They need the input to be converted into **tokens** and **embeddings**.

### **What Is Tokenization?**

Tokenization is the process of breaking text into small, meaningful pieces called **tokens**.

These tokens can be:

* words
* subwords (parts of words)
* characters

#### **Simple example (word-level)**

Text:
`"Data science is powerful"`

Tokens:
`["Data", "science", "is", "powerful"]`

#### Subword tokenization (intuitive idea)

Some languages and long words are broken into chunks so that rare words can still be understood:

* `"unhappiness"` → `"un"`, `"happy"`, `"ness"`

This lets models reuse pieces like `"happy"` across many words.

#### **Why tokenization is needed**

* Models cannot handle raw sentences directly.
* Tokenization gives them a **discrete sequence of items** to process.
* It is the first step before converting text into numerical form.

### **What Are Embeddings?**

Embeddings are **numerical vectors** that represent text, images, or other data in a way that captures meaning and relationships.

You can think of an embedding as placing each word or sentence as a point in a **high-dimensional space** where:

* similar things are close together
* dissimilar things are far apart

#### **Intuitive example (words)**

If we plot words in an imaginary “meaning space”:

* “king” and “queen” will be close
* “car” and “bus” will be close
* “car” and “banana” will be far apart

The numbers inside the embedding vector encode this closeness mathematically.

#### **What embeddings can represent**

* single words
* pieces of text (sentences, paragraphs, documents)
* images
* audio clips

Embeddings act as **powerful features** that capture deep properties of the input.

### **Why Machines Need Tokenization and Embeddings**

Computers understand only numbers, not text or images.
To make AI systems work with human data:

1. **Tokenization** breaks text into units the model can step through.
2. **Embeddings** convert those units into numerical vectors that carry meaning.

**Without these two steps:**

* a chatbot could not understand your question
* a search engine could not find similar documents
* a recommendation system could not match “you might also like…”
* a model could not complete sentences or translate text

**In other words:**

> **Tokenization and embeddings are the bridge between human content and machine understanding.**

## **Summary of a Data Pipeline**

A data pipeline is a sequence of steps that move data from **collection** to **analysis or modeling**.
Here is an overview of the major stages you learned.

image.png

### **1. Data Acquisition — Getting the Data**

This is where everything begins.
Data is collected from:

* files (CSV, Excel)
* databases
* APIs
* websites
* sensors
* user-generated content (text, photos, audio)

At this stage, data is usually raw, unstructured, and inconsistent.

### **2. Data Management - Organizing and Storing Data**

Once collected, data must be stored properly so it can be accessed, reused, and protected.

Key ideas include:

* structured vs unstructured storage
* folders, files, and databases
* metadata (data about data)
* privacy and anonymization
* ethical handling of user information

Good data management ensures the data stays safe, consistent, and clearly organized.

### **3. Data Cleaning - Fixing Problems in the Data**

Raw data almost always contains issues.
Cleaning removes or corrects:

* missing values
* duplicates
* incorrect formats
* inconsistent categories
* outliers
* noisy or low-quality content (especially in text and images)

Data cleaning is crucial because it directly affects the accuracy and reliability of later analysis.

### **4. Exploratory Data Analysis (EDA) - Understanding the Data**

EDA helps us deeply understand the dataset:

* what values are common or rare
* how data is distributed
* trends and patterns
* relationships between variables
* potential issues still hiding inside the data

Using summaries and visualizations (histograms, bar charts, line charts, scatter plots), we build a clear picture of what the data is telling us.

### **5. Feature Engineering - Preparing Data for Models**

Before a machine learning model can learn patterns, the data must be shaped into meaningful inputs.

Feature engineering includes:

* cleaning and standardizing categories
* handling numerical values
* creating new helpful features
* binning and grouping
* encoding categorical features
* preparing unstructured data (text, images)
* tokenization
* embeddings

This step helps the model learn faster, perform better, and make more accurate predictions.

## **A simple real-world example that combines all above discussed topics**

To make everything clear, let’s walk through a real-world scenario using all the topics we covered.


### **Scenario:**

**An online food delivery company wants to understand why some customers stop ordering (customer churn).**

The goal is to use data to analyze customer behavior and eventually predict who might stop ordering in the future.

### **Step 1: Data Acquisition**

The company gathers data from:

* order history (database)
* customer profiles (registration forms)
* app usage logs
* customer reviews (text)
* delivery images (photos uploaded by delivery partners)

At this point, data is scattered across systems and comes in many forms.

### **Step 2: Data Management**

The data is stored in a well-structured system:

* customer tables
* order tables
* delivery tables
* review text stored separately
* metadata describing each field

Sensitive information like phone numbers and addresses, it is anonymized to protect privacy.

### **Step 3: Data Cleaning**

The dataset contains issues such as:

* missing values in customer age
* duplicate customer IDs
* inconsistent city names (“Hyd”, “Hyderabad”, “hyd”)
* outlier order amounts (₹1,00,000 from a data error)
* extremely short or irrelevant review messages
* dark or blurry delivery images

These problems are fixed so the dataset becomes stable and reliable.

### **Step 4: Exploratory Data Analysis (EDA)**

Analysts explore the cleaned dataset:

* Plot histograms of order amounts
* Compare order counts across cities
* Look at month-wise trends in user activity
* Examine relationships: age vs number of orders
* Study word frequency in customer reviews
* Check image quality and patterns

EDA helps uncover patterns such as:

* users who stop ordering typically reduce activity 2–3 weeks beforehand
* negative reviews often mention “late delivery”
* a particular city has unusually high churn
* spending habits differ by age group

### **Step 5: Feature Engineering**

The team prepares features for a churn prediction model:

* numerical fixes (fill missing ages, correct formats)
* cleaned categories (standardized city names)
* new features like:

  * average order value
  * time since last order
  * number of late deliveries
  * sentiment score from reviews
  * order frequency per month
* encoded features (one-hot encoding for city and cuisine)
* tokenized review text
* embeddings representing customer review meaning
* basic image features (brightness, clarity scores)

Now the dataset is properly structured and information-rich.

### **Final Outcome**

By following the full data pipeline:

* the raw data becomes understandable
* patterns become visible
* meaningful features are created
* the company can build a solid machine learning model
* the model can identify customers likely to stop ordering
* the company can take action early (offers, support, engagement)

This is how all the topics you learned work together in a real project.

# **1. Introduction to Model Development**

Model development is the stage where we transform data into something useful — a machine that can **recognize patterns, make predictions, or even generate new content**.
All the previous work you learned in “Data Foundations” exists to support this step. Clean, organized data is the foundation on which models learn.

This section helps you understand:

* what a model is,
* how it learns, and
* what steps are involved in building one.

Once you understand this pipeline, everything else in AI will start to make sense.


## **What Does “Model Development” Mean?**

In simple terms, **model development is the process of building an intelligent system using data**.

It includes:

* choosing the right model for the task
* feeding it the correct type of data
* allowing it to learn patterns
* checking how well it learned
* improving it step-by-step

Unlike traditional programming where you write exact instructions, model development is about:

> **teaching the computer by showing examples, not writing rules.**

A model does not memorize your data.
It tries to **generalize** — meaning it should perform well on new data it has never seen before.

## **What Is a Model?**

A **model** is not magic. It is simply a mathematical system that has learned to map:

> **Input → Output**
> (based on patterns from the training data)

Because AI covers many fields, models come in different forms. Let’s break it down in ML, DL and GenAI POV.

image.png

### **1. Machine Learning Models (ML) POV:**

Machine Learning models learn patterns from structured data (tables with columns).
They find relationships like:

* “higher study hours → higher marks”
* “age and spending are related”
* “some keywords indicate spam emails”

ML models are good for:

* classification (spam or not spam)
* prediction (house price)
* clustering (customer grouping)
* forecasting (sales next month)

**Examples of ML models:**

* Linear Regression
* Logistic Regression
* Decision Trees
* Random Forest
* SVM

These models require some feature engineering and human guidance.

### **2. Deep Learning Models (DL) POV**

Deep Learning works well for unstructured data like:

* images
* audio
* text
* videos

Instead of manually designing features, DL models automatically learn complex patterns.

**Examples:**

* CNNs → image recognition
* RNNs / LSTMs → sequence prediction
* Transformers → language, multimodal tasks

Deep learning is powerful because it can capture fine details that humans cannot easily specify.

### **3. Generative AI Models (GenAI) POV**

GenAI models do not just classify or predict — they **create**.

They can:

* generate text
* write code
* summarize books
* produce images
* answer questions
* create music
* analyze multimodal inputs

Examples:

* GPT-style language models
* Diffusion models (image generation)
* Multimodal models (understand text + image together)

These models learn from enormous datasets and create new outputs by understanding patterns deeply.

## **How Models Learn Patterns from Data**

A model “learning” is actually a systematic process of:

* seeing many examples
* making guesses
* comparing guesses with the actual answers
* adjusting the internal parameters
* repeating this cycle thousands of times

This is known as **training**.

Let’s break it down in a more simple way.

### **How Learning Works (Human Analogy)**

Imagine teaching a child to identify animals:

1. You show many pictures of cats and dogs.
2. The child guesses wrongly at first.
3. You correct them (“No, that’s a dog”).
4. The child updates their understanding.
5. After enough corrections, they get good at it.

Models go through the same cycle:

* “See data → make prediction → get feedback → improve.”

### **How Learning Works (Machine Perspective)**

Internally, the process is:

* The model makes predictions using some numbers (parameters).
* It compares predictions with real answers (error).
* A mathematical process reduces the error step by step.
* Each iteration updates the parameters so predictions improve.

Beginners don’t need the math — just the idea:

> **A model improves by reducing its mistakes after each attempt.**

### **What Helps a Model Learn Better?**

* high-quality cleaned data
* properly engineered features
* correct model choice
* enough training examples
* balanced datasets
* meaningful feedback through evaluation metrics

Without these, even the best model fails.

## **Overview of the Model Development Pipeline**

Model development is not a single step.
It is a pipeline — a structured sequence of stages.

Below is the clear, beginner-friendly pipeline we’ll use throughout this module.

image.png

### **1. Input Data**

The cleaned, prepared data enters the process.
This includes:

* numerical data
* categorical data
* text
* images
* multimodal combinations

This is why the entire “Data Foundations” module was essential.

### **2. Model Selection**

Here we decide which model suits:

* the task (classification, generation, prediction)
* the data type (numbers, text, images)
* performance needs
* speed requirements

Good selection saves time and improves accuracy.

### **3. Training the Model**

Training is where learning happens.

Key ideas:

* The model sees large amounts of data.
* It tries to find patterns linking input to output.
* It keeps adjusting its internal parameters to reduce error.
* This happens over many cycles (epochs).

This is just like practicing a skill repeatedly until you improve.

### **4. Validation**

We check how the model performs on **unseen** data.
Validation helps us:

* avoid overfitting
* tune model choices
* adjust hyperparameters
* select the best version of the model

Think of validation as preparing students with mock exams.

### **5. Testing**

Testing is the final evaluation.
Here we measure how good the model truly is in real-world situations.

The test data:

* is never used during training
* gives an honest estimate of performance
* is used only at the end

### **6. Evaluation with Metrics**

Metrics give numerical answers to “How good is our model?”

Examples:

* accuracy, precision, F1-score (ML models)
* BLEU, ROUGE, FID, perplexity (GenAI models)

Metrics make model performance measurable and comparable.

### **7. Iteration & Improvement**

Model development is rarely “perfect on first try.”

We improve the model by:

* adding better features
* tuning parameters
* changing architectures
* cleaning data better
* tracking experiments
* trying alternative models

This loop continues until the model reaches acceptable performance.

## **2. Model Selection (Choosing the Right Model)**

Model selection is the stage where we carefully decide **which type of model** is most suitable for the problem we want to solve.

Just like a doctor chooses different treatments for different patients, data scientists choose different models depending on the nature of the task, the available data, and the final goal.

Choosing the right model is not about picking the “most advanced” one.

It’s about picking the one that fits the problem **best**. In many real projects, a simple model often performs better than a complex one because it matches the data more closely.

### **Why Model Selection Matters**

The model you choose can completely change the outcome of a project.
Even if your data is high-quality, a poor choice of model can lead to:

* inaccurate predictions,
* long training times,
* wasted resources, or
* confusing results.

A suitable model, on the other hand, can learn quickly, give reliable outputs, and be easier to maintain.

For example, using a simple Logistic Regression model may work perfectly for predicting whether a customer will cancel a subscription. Using a large deep-learning model for this task would be unnecessary and inefficient.

Model selection is therefore about **finding the right balance between performance, speed, and practicality**.

## **3. Model Training Basics**

Model training is the stage where an AI system actually *learns*.

Up to this point, all the work you’ve done—cleaning data, exploring patterns, preparing features—has been preparing the ground. Training is where the model studies examples, discovers relationships, and forms internal rules that allow it to make predictions.

Think of training as a learning journey: the model starts with no knowledge and gradually becomes smarter by repeatedly correcting its mistakes.

### **What Is Training?**

Training means giving a model many examples and allowing it to adjust itself based on how well it performs.

You don’t manually teach the model rules.
Instead, training lets the model find those rules by looking at data.

In simple terms:

> **Training = learning from examples + adjusting based on mistakes.**

The model sees inputs, tries to guess the outputs, checks how far it is from the correct answer, and improves little by little. After many rounds, it becomes good at the task—just like a person learning a new skill through practice.

Unlike programming where the developer writes exact instructions, training allows the model to “discover” patterns on its own.

### **How Models Learn Patterns**

Inside every model are many internal values called **parameters**.

These parameters determine how the model makes predictions.

At the beginning of training, these parameters are random.
So the model’s predictions are usually poor.

Through training:

1. The model makes a prediction.
2. It checks how wrong the prediction is.
3. It adjusts the parameters slightly.
4. It tries again with new data.
5. Over time, the adjustments accumulate, and the model gets better.

This cycle repeats thousands of times.

#### **An analogy**

Imagine learning to play basketball.
At first, your aim is random. But after every shot, your brain adjusts:

* too strong → reduce force
* too short → push harder
* wrong direction → change angle

Through repeated correction, you improve.
Models learn the same way—through repeated feedback.

The important idea is:

> **A model learns by gradually reducing its error.**

### **Why Training Data Matters So Much**

Training data is the model’s teacher.
Good data teaches the right patterns; poor data teaches the wrong ones.

A model trained on messy or biased data will always give unreliable predictions—even if the model itself is advanced.

#### **Good training data should be:**

* **Clean** → free from errors and duplicates
* **Correctly labeled** → especially in classification tasks
* **Representative** → reflects the real-world cases the model will face
* **Balanced** → avoids favoring one group unintentionally

For example, if 90% of your customer data comes from one city, the model may learn rules that work only for that city.

Or, if many training examples are mislabeled, the model will learn incorrect patterns.

That’s why you spent so much time in the earlier module preparing the data carefully.

### **Essential Training Concepts**

During training, you will encounter a few core terms that appear again and again—regardless of whether you’re working with ML, DL, or GenAI.
Understanding these concepts makes the entire process clearer.

Let’s break them down one by one.

#### **1. Epochs**

An **epoch** is one complete pass through the entire training dataset.

If your dataset has 10,000 samples:

* **1 epoch** = the model sees all 10,000 samples once
* **5 epochs** = the model sees the full dataset five times

**Why multiple epochs?**

Because the first time through the data, the model is still guessing.
With each pass, it refines its understanding.

**Analogy:**

Reading a textbook once gives you a basic idea.
Reading it multiple times helps you truly understand the content.


#### **2. Batch Size**

Instead of showing all data at once, we show the model *small portions* of data called **batches**.

Why?

* It reduces memory usage
* It makes learning more stable
* It speeds up training

If your dataset has 10,000 examples and your batch size is 100:

* Each epoch has 100 batches
* The model updates itself 100 times per epoch

Think of batching as studying in small chunks instead of trying to memorize an entire book in one go.

#### **3. Iterations**

An **iteration** is a single update step based on one batch of data.

So:

* **Iterations per epoch = number of batches per epoch**
* If you have 10,000 rows and batch size is 100 → 100 iterations per epoch

Iteration is simply the number of “learning steps” the model takes.

#### **4. Loss Function (Beginner-Friendly Explanation)**

A **loss function** tells the model *how wrong it is*.

After every prediction, the model gets a numerical score that represents error.
This score is its guide.

* High loss → far from the correct answer
* Low loss → closer to the correct answer

Training aims to **minimize loss**.
Every adjustment the model makes is designed to push the loss downward.

A simple way to think about loss:

> **Loss is like the model’s “mistake meter.”
> The goal is to make this meter as low as possible.**

### **Underfitting and Overfitting**


image.png

The aim of training is not just to perform well on training data—it’s to perform well on new data too.
Two common problems arise when this balance isn’t achieved:

#### **Underfitting**

Underfitting happens when the model is too simple and cannot capture the actual patterns in the data.

Symptoms:

* poor accuracy on training data
* poor accuracy on test data
* predictions feel “too basic” or incomplete

Example:
Trying to fit a straight line to data that clearly forms a curve.

Real-life analogy:

> A student who studied very little and fails both practice and final exams.

#### **Overfitting**

Overfitting happens when the model memorizes the training data too well.
It performs extremely well on the data it has seen but poorly on new examples.

Symptoms:

* very high training accuracy
* low test accuracy
* model reacts strongly to tiny details or noise

Example:
A student who memorizes past test papers word-for-word but fails when given slightly different questions.

### **What We Really Want: Generalization**

A good model:

* learns real patterns
* ignores noise
* performs well on unseen data

Generalization is the goal of all training — it means the model has truly *learned*, not memorized.

## **4. Data Splits: Training, Validation & Test Sets**

When developing a model, one of the most important principles is ensuring that it learns patterns that generalize to new, unseen data.
To achieve this, we **never train and evaluate the model on the same data**. Instead, we divide the dataset into separate parts, each with a specific purpose.

This process of splitting data prevents the model from “cheating,” protects against bias, and helps us measure true performance.

### **Why Do We Split Data? (Avoiding Cheating & Bias)**

If we trained and tested the model on the same dataset, the model could achieve extremely high accuracy simply because it has already seen the answers. This is similar to giving a student the exam answer key before the test—they will score perfectly, but they haven’t truly learned the subject.

#### **Splitting data helps us:**

* **Check real learning:** The model must perform well on data it has never seen before.
* **Prevent overfitting:** The model cannot simply memorize the training examples.
* **Detect problems early:** Poor performance on validation/test sets warns us that the model isn’t generalizing.
* **Make fair comparisons:** Different models can be evaluated using the same test set.

In short:

> **Data splitting is essential to ensure the model genuinely learns patterns instead of memorizing answers.**

### **Purpose of Each Split**

We typically divide data into **three sets**, each with a specific role.

1. Training set
2. Testing set
3. Validation set

#### **1. Training Set — “Learning Phase”**

The training set is the portion of data the model **directly uses to learn patterns**.

During training:

* the model sees the inputs,
* compares predictions with actual outputs,
* adjusts its parameters based on the error,
* and repeats this process many times.

This is like a student practicing with many exercises to understand the topic.

A good training set must be:

* large enough
* clean
* diverse
* representative of the real world

because the model builds its entire understanding from this portion.

#### **2. Validation Set — “Tuning & Decision Phase”**

The validation set acts like a **mock exam**.
It contains data the model has *not* seen during training.

We use it to:

* check how well the model is improving
* choose between different models
* tune hyperparameters (settings like learning rate, depth, layers)
* detect overfitting early
* decide when to stop training

The validation set helps us answer questions like:

* Has the model learned too much (overfitting)?
* Is the model learning too little (underfitting)?
* Which model performs best?
* Should we change settings to reduce error?

It is never used for final evaluation—only for making decisions during development.

#### **3. Test Set — “Final Performance Check”**

The test set is kept untouched until the very end.
It serves as the **final exam** for the model.

We use it to measure:

* how well the model will perform in real-world situations
* whether the model generalizes correctly
* whether we can trust the model beyond the training environment

If the model performs well on the training set and validation set but poorly on the test set, it means:

* our validation process was flawed, or
* the model has overfitted to both training and validation data.

This is why the test set must remain isolated until the development is complete.

### **Common Data Split Ratios**

There is no single “correct” split ratio.
The ideal ratio depends on dataset size, complexity, and the task.
However, beginners often use the following standard splits:

#### **70 / 20 / 10 (Training / Validation / Test)**

* 70% of the data is used for training
* 20% for validation
* 10% for testing

A balanced option when dataset size is moderate.

#### **80 / 10 / 10**

* 80% training
* 10% validation
* 10% test

Useful when your dataset is slightly smaller and you want to maximize training examples.


#### **How to choose? (Beginner rule of thumb)**

* **Large datasets** → smaller test/validation sets are fine
* **Small datasets** → keep test/validation sets small but meaningful
* **GenAI training** → often uses huge training sets and a tiny validation set

The goal is to strike a balance between learning and evaluating.

### **Cross-Validation**

Cross-validation is a method used when we want to make better use of our training data, especially when we have **limited data**.

Instead of having a single validation set, we rotate the validation role among different slices of the data.

#### **Simplest way to understand it:**

Imagine you have 5 slices of data: A, B, C, D, E.

* In round 1 → train on B, C, D, E and validate on A
* In round 2 → train on A, C, D, E and validate on B
* In round 3 → train on A, B, D, E and validate on C
* … and so on

This gives us:

* more reliable performance estimates
* better use of limited data
* protection against biased splits

Cross-validation is commonly used in ML competitions and projects where dataset size is limited.

### **Why Using the Test Set Too Early Is Dangerous**

A beginner’s mistake is checking the test set repeatedly while tuning the model.
This feels helpful, but it causes the model to accidentally “learn” from the test set.

This is a sneaky form of overfitting.

#### **Reason:**

If you tune the model based on test set performance:

* the test set stops being a fair measure
* the model is no longer being evaluated on truly unseen data
* performance metrics become misleading or inflated
* the model may fail in real-world use

#### **Think of it like this:**

> If a student sees the final exam questions before the test, their score becomes meaningless.

To keep the test score honest:

* **Never** use the test set during training
* **Never** tune hyperparameters using the test set
* Use the test set only once — at the end

This ensures we can trust the final evaluation.

## **Model Evaluation**

Once a model has been trained, the next question is:
**“How good is it?”**

Training alone does not guarantee usefulness. A model may learn too little, learn the wrong patterns, or even memorize the training data. Evaluation is the process that helps us measure the model’s real ability and decide whether it is ready for real-world use.

Model evaluation is not about guessing or subjective judgment.

It is about using clear, objective measurements to understand how well the model performs on data it has *never seen before*.

### **What Evaluation Means**

Evaluation is the step where we test the model’s performance and measure whether it truly learned meaningful patterns.

At this stage, we check:

* how accurately it predicts
* how consistently it performs
* whether it generalizes to new situations
* whether it makes certain types of mistakes
* whether it is reliable enough to be used in real-world applications

In simple words:

> **Evaluation tells us whether our model is trustworthy.**

Without evaluation, we would have no idea if the model is actually good or just appears good inside the training loop.

### **Why Evaluation Is Different for Different Tasks**

Not all tasks can be judged using the same standards.
Different types of problems require different ways of measuring success.

1. Classification tasks
2. Regression tasks
3. Clustering tasks
4. Deep learning tasks
5. GenAi tasks

#### **1. Classification Tasks**

When the output is a label (e.g., “spam” vs “not spam”), we care about:

* how many predictions were correct
* how often the model misses important cases
* how often it raises false alarms

Metrics like accuracy, precision, recall, and F1-score are used because they show how well the model handles different kinds of mistakes.

#### **2. Regression Tasks**

When the output is a number (e.g., house price or temperature), we look at:

* how far predictions are from real values
* how large the typical error is
* whether errors are consistent across ranges

Metrics like MAE, MSE, and RMSE are used to capture numerical differences.

#### **3. Clustering Tasks**

Clustering has no labeled outputs.
We often measure:

* how well groups are separated
* whether similar items are grouped together
* whether different groups overlap too much

Clustering evaluation uses different techniques like silhouette scores or visual inspection.

#### **4. Deep Learning Tasks (images, audio, sequences)**

For complex tasks such as:

* object detection
* speech recognition
* sentiment analysis
* image classification

We need metrics suitable for these domains.
Some tasks even require domain-specific accuracy measures (e.g., IoU for object detection).

#### **5. Generative AI Tasks**

Evaluation becomes even more unique for GenAI because the output is not a label or number — it is **created content**.

For example:

* text generation: summaries, translations, responses
* image generation: realistic or high-quality images
* multimodal tasks: describing images, answering questions

These tasks require metrics such as BLEU, ROUGE, CIDEr, FID, and perplexity.
These metrics compare generated content with human-created content to judge:

* quality
* relevance
* fluency
* similarity
* creativity
* correctness

This is why GenAI evaluation is more complex than traditional ML evaluation.

### **Overview of Evaluation in ML, DL, and GenAI**

Different model families use different evaluation strategies depending on the nature of their outputs.

#### **1. Machine Learning (ML) Evaluation**

ML tasks often involve structured data with clear, interpretable outputs:

* classification → accuracy, precision, recall, F1
* regression → MAE, MSE, RMSE
* clustering → silhouette score

ML evaluation tends to be more direct and numerical.

#### **2. Deep Learning (DL) Evaluation**

Deep learning models handle more complex data types like images, text, speech, and sequences.
Evaluation often includes:

* overall accuracy
* confusion matrices
* domain-specific metrics (IoU, BLEU for NLP, CER/WER for speech)
* visual inspection for tasks like image detection

DL evaluation also considers robustness — how well the model handles noisy or unseen variations.

#### **3. Generative AI (GenAI) Evaluation**

GenAI systems require the most nuanced evaluation because the output is **creative** rather than fixed.

For text models:

* BLEU, ROUGE → compare with reference text
* Perplexity → measures how “surprised” a model is

For image generation:

* FID → compares real vs generated image distributions
* IS (Inception Score) → checks image quality and diversity

Sometimes, human evaluation is needed, especially for subjective tasks such as:

* rating helpfulness
* judging creativity
* assessing factual accuracy
* determining safety and appropriateness

### **Why Evaluation Matters Before Deployment**

A model that looks perfect during training may fail in the real world if:

* the evaluation step was skipped
* the wrong metric was chosen
* the test data was not representative
* the evaluation was done using biased samples

Evaluating a model correctly ensures:

* fairness
* reliability
* safety
* performance stability
* trustworthiness

A model is only ready for deployment when it consistently performs well on **unseen, diverse, real-world-like** data.

## **ML Metrics (Classification & Regression)**

Model evaluation is not complete without proper **metrics**. Metrics help us measure the quality of a model’s predictions in a way that is objective, consistent, and comparable.
Without metrics, we cannot tell whether one model is better than another, or whether the model is ready for real-world use.

For beginners, the key point is:

> **Different problems require different types of metrics.**
> Classification uses one set of metrics, and regression uses another.

This section explains the most common metrics used in traditional machine learning.

#### **Why Metrics Matter**

During training, a model may seem to perform well.
But models can be misleading:

* They may get many predictions correct but fail at important cases.
* They may perform well on average but poorly on extreme values.
* They may score high on one metric but low on another.

Metrics give us a **complete picture** of performance, helping us answer questions like:

* Is the model good at catching positive cases?
* Is the model avoiding false alarms?
* Are the predicted numbers close to actual values?
* Is the model making systematic mistakes?

Metrics reveal strengths and weaknesses that accuracy alone cannot show.

### **A. Classification Metrics**

Classification metrics are used when the output is a **category** (e.g., spam/not spam, churn/no churn, dog/cat).

Let’s explore the most important ones.


#### **1. Accuracy**

Accuracy measures **how many predictions were correct** out of all predictions.

It is simple and intuitive:

> **Accuracy = “What percentage of the time was the model right?”**

#### **When Accuracy Is Useful**

* When classes are balanced
* When mistakes have similar consequences

#### **When Accuracy Is Misleading**

Accuracy can hide problems when the dataset is **imbalanced**.

For example:
If 98% of transactions are normal and only 2% are fraud:

* A model that predicts “normal” for everything will get 98% accuracy
* But it catches **zero** fraud cases

So the accuracy looks high, but the model is useless.

This is why we need precision, recall, and F1-score.

#### **2. Precision**

Precision answers the question:

> **“Out of everything the model predicted as positive, how many were actually positive?”**

Example:
A spam detector flags 100 emails as spam.
If 80 of them truly are spam → precision = 80%.

#### **Useful when:**

* False alarms are costly
* E.g., classifying legitimate emails as spam is bad
* E.g., wrongly flagging a non-fraud transaction as fraud causes inconvenience

High precision means **few false positives**.


#### **3. Recall**

Recall answers the question:

> **“Out of all the actual positive cases, how many did the model correctly detect?”**

Example:
There are 100 real spam emails.
If your model catches 70 → recall = 70%.

#### **Useful when:**

* Missing positive cases is dangerous
* E.g., failing to detect cancer
* E.g., failing to catch fraud
* E.g., detecting safety issues

High recall means **few false negatives**.

#### **4. F1-Score**

Precision and recall often conflict — improving one may reduce the other.
The F1-score gives a **balanced measure** between both.

It is especially useful when:

* classes are imbalanced
* both false positives and false negatives matter
* accuracy alone is misleading

Think of F1-score as:

> **A single score that combines precision and recall fairly.**

The higher the F1-score, the better the model balances both sides.

#### **5. Confusion Matrix (Intuitive Explanation)**

A confusion matrix is a small table that shows where the model was right and where it was wrong.

It has four components:

|                     | **Predicted Positive** | **Predicted Negative** |
| ------------------- | ---------------------- | ---------------------- |
| **Actual Positive** | True Positive (TP)     | False Negative (FN)    |
| **Actual Negative** | False Positive (FP)    | True Negative (TN)     |

#### **Why it helps**

Instead of a single number like accuracy, the confusion matrix shows:

* where the model confuses classes
* how many false alarms (FP) it causes
* how many important cases it misses (FN)
* whether the model is biased toward one class

It gives a full picture of model behavior.

### **B. Regression Metrics**

Regression metrics are used when the output is a **number**, such as price, temperature, or sales.

These metrics measure **how far the predicted value is from the actual value**.

#### **1. MAE — Mean Absolute Error**

MAE tells us the **average size of mistakes**, without considering direction.

If the model predicts:

* 110 instead of 100 → error = 10
* 90 instead of 100 → error = 10

Both errors are treated equally.

#### **Why MAE is useful**

* Easy to understand
* Treats all errors fairly
* Good when large mistakes are acceptable

#### **2. MSE — Mean Squared Error**

MSE also measures the size of errors, but it **squares** them.

This means:

* larger mistakes receive much bigger penalties
* the model becomes extra sensitive to big errors

Example:

* error of 10 → contributes 100
* error of 20 → contributes 400

#### **Why MSE is useful**

* Encourages the model to avoid large errors
* Helpful when big mistakes are expensive

#### **3. RMSE — Root Mean Squared Error**

RMSE is simply the square root of MSE.

It has the advantage that:

* its units match the original target
* it gives more weight to large errors
* it is considered one of the best regression metrics for general use

#### **When to use RMSE**

* When outliers matter
* When large mistakes must be minimized
* When comparing models on real-valued scales

### **Why Accuracy Can Be Misleading (Imbalanced Classes)**

One of the most important lessons in machine learning is:

> **High accuracy does not always mean a good model.**

Consider a medical dataset where only 1% of cases are positive (disease present).
If a model predicts “no disease” for everyone:

* Accuracy = 99%
* Recall = 0% (it caught zero positive cases)
* F1-score = terrible

This model is useless even though accuracy looks perfect.

#### **Why this happens**

Accuracy treats all errors equally, but in reality:

* Missing a fraud case is worse than wrongly flagging a normal one
* Missing cancer is far more serious than a false alarm

So **precision, recall, and F1-score** are essential when classes are imbalanced.

## **C. GenAI-Specific Metrics**

Evaluating Generative AI is very different from evaluating traditional machine learning models.
Unlike classification or regression tasks—where there is a single correct answer—Generative AI produces **open-ended outputs** such as:

* summaries
* translations
* full paragraphs
* images
* captions
* music

Because the outputs are creative and varied, we cannot judge them using standard metrics like accuracy or MSE.
Generative AI requires **special metrics** designed to measure *quality*, *similarity*, *coherence*, and *faithfulness*.

### **Why GenAI Needs Different Metrics**

In classical ML, there is usually a clear right or wrong answer.
Example: Email is either spam or not spam.

But in GenAI:

* A model can write multiple valid summaries of the same article.
* A translated sentence can be phrased in many correct ways.
* Two images can both be “high quality,” yet look completely different.
* Creativity cannot be measured with one strict answer key.

Therefore:

> **GenAI evaluation focuses on how close, meaningful, and high-quality the generated output is compared to human-created examples.**

We evaluate GenAI based on:

* quality of text or images
* similarity to reference outputs
* fluency and coherence
* diversity
* relevance
* realism

The metrics below help measure these qualities.

### **D. Text Generation Metrics**

Text generation tasks include:

* translation
* summarization
* chatbot responses
* question answering
* content creation

For these tasks, we evaluate how close the model’s output is to human writing.

Let’s explore the main text metrics.

#### **1. BLEU (Bilingual Evaluation Understudy)**

BLEU is one of the earliest and most widely used metrics for evaluating **machine translation** and **text generation**.

#### **What BLEU measures**

BLEU checks how many words or short phrases in the generated text match the words/phrases in the reference text.

#### **Intuition**

If the model output uses similar wording as the reference, BLEU score will be high.

Example:

Reference: *“The cat sits on the mat.”*

Generated: *“A cat is sitting on the mat.”*

The phrases “cat”, “on the mat”, “sitting/sits” match → good BLEU.

#### **What BLEU does NOT measure**

* meaning
* creativity
* clarity
* long-range coherence

So BLEU is helpful but limited.

#### **2. ROUGE (Recall-Oriented Understudy for Gisting Evaluation)**

ROUGE is mostly used for **summarization** tasks.

#### **What ROUGE measures**

ROUGE checks how much of the important content from the reference text appears in the generated summary.

BLEU looks at overlap from the system’s perspective.
ROUGE looks from the reference’s perspective.

#### **Intuition**

If the model summary captures the key ideas from the original summary, ROUGE score is high.

#### **Why this matters**

A good summary should:

* capture main ideas
* include important keywords
* avoid unnecessary details

ROUGE does a decent job of measuring these qualities.

#### **3. Perplexity**

Perplexity measures how “surprised” a language model is by the next word in a sentence.

#### **Intuition**

A good model predicts the next word with high confidence → **low perplexity**.
A weak model is confused or uncertain → **high perplexity**.

Example:

Sentence: *“The sun rises in the ___.”*

A strong model expects “east” with high confidence → low perplexity.

A confused model might consider “east, west, morning…” → high perplexity.

#### **Where perplexity is used**

* training language models
* comparing model versions
* checking if a model understands language patterns well

#### **Limitation**

Low perplexity does not guarantee good writing quality.
The text may be predictable but boring or repetitive.

### **E. Image Generation Metrics**

Evaluating images is harder than evaluating text because there is no “single correct image.”
We focus on whether the generated images are:

* realistic
* diverse
* similar to real images
* visually high-quality

Two main metrics help here: **FID** and **Inception Score**.

#### **1. FID (Fréchet Inception Distance)**

FID is the most widely used metric for evaluating image generators like diffusion models.

#### **What FID measures**

FID compares two sets of images:

* real images
* generated images

It checks:

* how similar the overall patterns are
* how close the distributions of features are
* how realistic the generated images look

#### **Low FID = better quality**

Low FID → generated images are similar to real images
High FID → generated images are poor or unrealistic

#### **What FID captures**

* sharpness
* diversity
* realism

This makes FID the gold standard for image generation evaluation.

#### **2. Inception Score (Optional but Simple)**

The Inception Score evaluates two things:

1. **Image clarity**
   — Are the generated images recognizable and meaningful?
2. **Image diversity**
   — Are the images varied, or does the model repeat itself?

#### **Interpretation**

Higher IS → better quality and diversity.

However, IS has limitations and is used less frequently today compared to FID.

### **What These Metrics Try to Measure**

Although these metrics differ in design, they aim to measure similar qualities:

#### **For Text**

* **Similarity** to human-written content
* **Coherence** — does the text flow logically?
* **Relevance** — does it stay on topic?
* **Fluency** — does it read naturally?

#### **For Images**

* **Realism** — does it look like a real image?
* **Sharpness** — is it clear or blurry?
* **Diversity** — does the model produce varied outputs?
* **Faithfulness** — does the image match the prompt?

#### **Why these features matter**

A generative model is not just judged on correctness, but on how well it **creates**.

This makes GenAI evaluation more subjective and complex than traditional ML evaluation.

## **Iteration, Improvement & Experiment Tracking**

Model development is **never a one-shot activity**.
Even experienced data scientists do not build a perfect model on their first attempt.
Instead, model building is an *iterative* process — meaning we continually try, evaluate, adjust, and improve.

Think of it like refining a recipe.
You don’t get the perfect taste on the first try.
You test different ingredients, adjust portions, and keep notes until you find what works best.

In machine learning and GenAI, this process is exactly the same.

### **What “Iteration” Means in ML Projects**

Iteration means **repeating the cycle of training → evaluating → improving** until the model reaches acceptable performance.

Each iteration is like a new experiment:

* You change one element (a feature, a parameter, or data balance).
* You train again.
* You measure performance.
* You compare with earlier results.

Over time, these small improvements can dramatically upgrade model quality.

#### **Why iteration is necessary**

* The first model is rarely the best.
* Data may have hidden patterns the initial model can’t capture.
* You discover weaknesses only after evaluation.
* Different models behave differently on the same dataset.

So **iteration is how we evolve a model from “okay” to “excellent.”**

### **Steps Involved in Improving a Model**

image.png

Improvement does not mean “guessing randomly.”

There are structured ways to make a model better — let’s explore them with clear explanations for beginners.

1. Try different features
2. Tune hyperparameter
3. Change model architecture
4. Collect more and better data
5. Balance the dataset.

#### **1. Try Different Features**

Features are the model’s “inputs.”
Poor or missing features lead to poor predictions.

Improving features can mean:

* creating new useful columns
* correcting or simplifying categories
* grouping values
* normalizing numerical ranges
* removing noisy or irrelevant features

#### Example

Instead of using raw “age,” using “age group” (0–18, 18–35, 35–60, 60+) might capture patterns more clearly.

Good features often matter **more** than choosing a fancy model.


#### **2. Tune Hyperparameters**

Hyperparameters are *settings* that control how the model learns.
Examples include:

* learning rate
* number of layers
* number of trees in a random forest
* batch size
* training epochs

Small changes in these settings can significantly affect:

* how fast the model learns
* whether it overfits
* whether it converges

Hyperparameter tuning is like adjusting oven temperature or cooking time — the recipe stays the same, but the outcome can improve drastically.

#### **3. Change the Model Architecture (DL/GenAI)**

For deep learning and generative AI, the “architecture” is the structure of the model.
Sometimes performance improves by:

* adding/removing layers
* increasing network depth or width
* using different types of layers
* switching from RNN to Transformer
* trying a different loss function
* using pre-trained models instead of training from scratch

Architecture changes are more advanced, but beginners should know that:

> **Deep learning models can be reshaped like building blocks.
> Different shapes capture different patterns.**

#### **4. Collect More or Better Data**

If your model struggles to generalize, it may not have enough examples.

More data helps the model:

* see more variations
* learn more stable patterns
* reduce overfitting
* become more confident

Sometimes the issue is not the amount but the **quality** of data.

Better data may include:

* fixed labels
* more balanced examples
* updated datasets
* removing mistakes or duplicates

High-quality data often improves performance more than changing the model.

#### **5. Balance the Dataset**

Many real-world datasets are **imbalanced** — one class has far more examples than another.

For example:
98% non-fraud, 2% fraud.

In such cases, the model may learn to ignore the minority class.

Balancing can involve:

* oversampling (adding more minority samples)
* undersampling (reducing majority samples)
* using class weights
* generating synthetic samples (SMOTE)

Balanced data helps the model learn all classes fairly.

### **What Is Experiment Tracking?**

When you try many variations — different features, different hyperparameters, different models — it becomes difficult to remember what you changed and which version worked best.

This is where **experiment tracking** becomes essential.

#### **Experiment Tracking = Keeping a Clear Record of Every Model Version**

It allows you to:

* save each experiment’s settings
* log the metrics achieved
* track which version performed best
* avoid repeating mistakes
* understand why a model improved
* share results with teammates

Think of experiment tracking as keeping a scientific lab notebook.

### **Why Tracking Matters**

#### **1. Version Control for Models**

Just like software has versions, models do too.
Tracking helps you know:

* which version is currently deployed
* which version performed better
* what changes were made between versions

Without tracking, it's easy to lose track of progress.

#### **2. Logging Metrics for Comparison**

Each experiment produces important metrics:

* accuracy
* F1-score
* BLEU
* FID
* training time
* loss curves

Experiment tracking organizes all this so you can compare experiments side-by-side and choose the best one.

#### **3. Reproducibility**

Reproducibility means another person — or even *you* after six months — can repeat the same experiment and get the same results.

Tracking ensures:

* the exact settings are saved
* the data version is known
* the model configuration is recorded

This is critical for professional AI development.

#### Simple analogy:

> If a scientist discovers something but forgets the steps, the discovery becomes useless.
> Tracking prevents this problem in model development.

### **Tools for Experiment Tracking**

Machine learning projects rarely succeed in a single attempt. Models are trained, evaluated, adjusted, and retrained multiple times. This process is called **iteration**.

As experiments increase, it becomes difficult to remember:

* Which model version performed best
* Which parameters were used
* Which dataset produced which result

This is where **experiment tracking tools** are used.

Experiment tracking tools ensure that:

## **Simple Real-World Walkthrough**

image.png

Model development can feel abstract until you see how everything connects in a realistic scenario.
In this walkthrough, we will step through a simple, beginner-friendly example so students can visualize the *entire* workflow from model selection to improvement.

Let’s imagine we work for a telecom company that wants to predict **customer churn** — which customers are likely to stop using their service. The goal is to identify at-risk customers early and offer them support.

This problem is straightforward, uses structured data, and connects naturally to the concepts students have learned.

#### **Step 1: Understanding the Problem**

The first step is recognizing what kind of task this is. Predicting whether a customer will churn is a **classification problem**, because the outcome is a category — *yes* or *no*.
Students should notice that classification tasks often include real-world variables such as customer service calls, monthly bills, contract types, usage patterns, and more.

At this stage, no model is built yet. We simply understand what kind of prediction we need to make.

#### **Step 2: Selecting a Suitable Model**

Since the data consists of columns and rows — typical structured data — it makes sense to begin with a simple and interpretable model. Logistic Regression is a popular choice for binary classification and gives quick, reliable results.

A Decision Tree is another option if we want to visualize how decisions are made.

Deep learning would be unnecessary here. The dataset is not extremely large, the task is not complex, and interpretability matters.

Starting with a simple model allows us to build a baseline understanding before trying more complex approaches.

#### **Step 3: Splitting the Data**

To ensure the model learns genuinely and not by memorizing answers, we split the data into:

* a **training set**, where the model learns patterns
* a **validation set**, where we fine-tune the model
* a **test set**, which we keep untouched until the very end

A common split is 70% for training, 20% for validation, and 10% for testing.

This separation is essential because it prevents the model from cheating — performing well simply because it has already seen the data.

#### **Step 4: Training the Model**

Training is where the model finally starts learning. It studies training examples one batch at a time, compares its predictions with the true labels, and adjusts its internal parameters to reduce error.

During this process, the model gradually learns relationships such as:

* customers with frequent service complaints churn more
* customers with long-term contracts churn less
* payment method or usage patterns may influence churn

Training runs in **epochs**, meaning the model sees the entire dataset multiple times. Each round improves its understanding slightly.

#### **Step 5: Evaluating the Model**

Once training is complete, we check performance on the **validation set**.
For classification problems, accuracy alone is not enough. A high accuracy might hide the fact that the model is missing many customers who actually churn.

This is where **precision**, **recall**, and the **F1-score** become valuable:

* Precision tells us whether the model’s positive predictions are trustworthy.
* Recall tells us whether it can successfully identify most churn cases.
* The F1-score balances both and gives a single interpretable summary.

The **confusion matrix** helps us see exactly where mistakes occur — such as the model predicting “not churn” even when a customer actually churns.
This step guides our next decisions.

#### **Step 6: Improving the Model (Iteration)**

Rarely does a model perform perfectly on the first attempt.
Suppose the validation results show low recall — the model is missing too many actual churn cases. Now we begin the iterative improvement phase.

We might start by improving the **features**.

* Adding variables such as average monthly usage or time since last plan upgrade can help the model capture deeper behavior patterns.

* Next, we might **tune hyperparameters**, adjusting elements like regularization strength or decision thresholds to improve recall without sacrificing too much precision.

* If the simple model reaches its limits, we may try a different model such as a **Random Forest**, which captures more complex relationships.

* Finally, if the dataset is imbalanced — meaning few customers churn — we may improve balance using oversampling or class weights so the model pays more attention to the minority class.

* Each improvement forms one iteration. And each iteration brings the model closer to real-world usefulness.

#### **Step 7: Final Testing**

Once we settle on the best-performing version, we evaluate it on the **test set** — data the model has never seen during any stage.

This final score reflects how the model will behave in the real world. A strong test performance indicates that our modeling workflow was successful.

This completes the journey from understanding the problem to deploying a trustworthy model.

## **Summary of Model Development & Evaluation**

To tie everything together, it’s helpful to reflect on the full process of model development. Although each individual step is important, the real power comes from understanding how all the pieces connect.

* Model development begins with **choosing the right model**, which depends on the type of problem, the data involved, and the goals of the application.

* Once a model is selected, the dataset must be **properly split** so that training, validation, and testing serve their distinct roles. This ensures learning happens fairly and performance is evaluated honestly.

* During **training**, the model learns from repeated exposure to the data.
  
  It adjusts its parameters gradually, guided by the loss function. Beginners should appreciate that models improve through repetition, corrections, and exposure to diverse patterns.

* After training, **evaluation** becomes the spotlight. Metrics such as accuracy, precision, recall, MAE, BLEU, and FID help us measure success depending on the task.

  Different tasks require different metrics, because a single score cannot capture all success criteria.

  What follows is **iteration**. Improvement is natural — and expected.

* Real-world modeling is a cycle of revising features, tuning hyperparameters, adjusting architectures, fixing data issues, and experimenting with alternatives.

  Experiment tracking keeps this process organized, allowing data teams to record what was tried, what worked, and what didn’t.

Once we are satisfied with performance, we evaluate the final model on untouched test data and decide whether the model is ready for practical use.

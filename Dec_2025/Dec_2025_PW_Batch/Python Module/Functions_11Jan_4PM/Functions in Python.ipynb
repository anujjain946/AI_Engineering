*args: It collects the input in the form of tuple (like a list which is immutable). We use this in a situation where we have multiple values in the arguments.

````python
# function definition
def learners_info(*learners):
  print("The fifth learner enrolled in my Data Science course is: " + learners[4])


# function calling
learners_info("Swati", "Vikram", "Abhishek", "Khushboo", "Riya", "Rajesh", "Shyam", "Rashmi", "Priya", "Yameen")
````

**kwargs: used for keyword inputs - it collect them into a dictionary of key-value pairs.

````python
# function definition
def learners_info(**kwargs):
  for key, value in kwargs.items():
    print(key + ": " + value)

# function calling
learners_info(name="Swati", age="25", city="bengaluru")
````

Palindromic String

````python
# function definition
def is_palindrome(s):
    # comparing the original string with the reverse of the original string
    return s == s[::-1]

# function call
string = input("Enter your string: ")

if is_palindrome(string):
    print(f"'{string}' is a palindrome.")
else:
    print(f"'{string}' is not a palindrome.")
````

Recursion: A recursive function is a function that solves a problem by calling itself, breaking down complex tasks into smaller, identical subproblems until a simple "base case" is reached, which stops the recursion, preventing infinite loops.

````python
# function definition
def factorial(n):
  if n == 0 or n == 1:
    return 1
  else:
    # recursive function call
    return n * factorial(n - 1)

# function calling
n = 5

result = factorial(n)
print("The factorial of a given number is: ", result)
````

Scope of the variables:

* Global Scope

* Local Scope

Global Scope: Variables defined outside any function are considered to be in the global scope. These variables can be accessed from anywhere in the program, including within functions.

````python
x = 10 # global variable

def my_function():
  print(x)

my_function()
print(x)
````

Local Scope: Variables defined within a function are considered to be in the local scope. These variables are only accessible within the function where they are defined.

````python
def my_function():
  y = 20 # local variable
  print(y)

my_function()
print(y)
````

````python
x = "awesome" # global variable

def my_function():
  x = "interesting" # local variable
  print("Today's class is " + x)

my_function()
print("Today's class is " + x)
````

Nested functions: you are defining one function inside another function.

````python
def outer_function(x, y):
  def add(a, b):
    return a + b

  def multiply(a, b):
    return a * b

  def calculate(operation_type, a, b):
    if operation_type == "add":
      return add(a, b)
    elif operation_type == "multiply":
      return multiply(a, b)
    else:
      return "invalid operation"
  # function call
  return calculate("add", x, y), calculate("multiply", x, y)

# function calling
result = outer_function(3, 4)
print(result)
````

Lambda Functions: Anonymous functions which are useful only when we are doing some small task.

```python
lambda arguments: expression
```

````python
add = lambda x, y: x + y
print(add(3, 5))
````

````python
square = lambda num : num ** 2
print(square(5))
````

````python
multiply = lambda x, y: x * y
print(multiply(2, 3))
````

Map Function: to transform every element in the list

````python
salaries = [50000, 60000, 70000]

# mutiply all the values of original salaries by 1.10 [10%]
updated_salaries = list(map(lambda x : x * 1.10, salaries))

print(updated_salaries)
````

Filter Function: select only elements that satisfies the condition

````python
experience_years = [0, 1, 2, 3, 4, 5, 6]

eligible_candidates = list(filter(lambda x : x > 2, experience_years))

print(eligible_candidates)
````

Reduce Function: combines all the elements into a single value

````python
from functools import reduce

salaries = [50000, 60000, 70000]

total_salaries = reduce(lambda a, b: a + b, salaries)
print(total_salaries)
````

Import all the required libraries

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset("tips")
tips.head()
```

Average Tip By Day

```python
avg_tip_by_day = tips.groupby("day")["tip"].mean()
print(avg_tip_by_day)
```

Visualization

```python
plt.figure(figsize=(6, 4))
plt.plot(avg_tip_by_day.index, avg_tip_by_day.values, marker='o')
plt.title('Average tip by day')
plt.xlabel('Day')
plt.ylabel('Average tip')
plt.show()
```

Seaborn

```python
plt.figure(figsize=(6, 4))
sns.barplot(data=tips, x="day", y="tip")
plt.title("Average Tip by day")
plt.show()
```

Distributions on tips data

```python
plt.figure(figsize=(6, 4))
sns.histplot(data=tips, x="total_bill", bins=20, kde=True)
plt.title("Distribution of total bill")
plt.show()
```

BoxPlot -> majorly used to detect the outliers in a given data. All the circle on the above whisker and lower whisker are a part of the outlier data points.

```python
plt.figure(figsize=(6, 4))
sns.boxplot(data=tips, x="day", y="total_bill")
plt.title("Total Bill by day (boxplot)")
plt.show()
```

Create a boxplot of tip by smoker (Yes/No)

```python
plt.figure(figsize=(6,4))
sns.boxplot(data=tips, x="smoker", y="tip")
plt.title("Tip by smoker")
plt.show()
```

Relationship among total_bill vs tip column

```python
plt.figure(figsize=(6, 4))
sns.scatterplot(data=tips, x='total_bill', y='tip')
plt.title("Total Bill vs Tip")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex')
plt.title("Total Bill vs Tip")
plt.show()
```

Regression Plot

```python
plt.figure(figsize=(6, 4))
sns.regplot(data=tips, x="total_bill", y="tip")
plt.title("Trend")
plt.show()
```

Correlation Heatmap

```python
tips_subset = tips[['total_bill', 'tip', 'size']]

corr = tips_subset.corr()
print(corr)
```

```python
sns.heatmap(corr, annot=True)
plt.title("Correlation Matrix")
plt.show()
```

Countplot of smoker by time (Lunch/Dinner).

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=tips, x="time", hue='smoker')
plt.xlabel('Time')
plt.ylabel('Count')
plt.show()
```

Pairplot

Insights: Strongest positive correlation between total bill and tip value.

```python
sns.pairplot(tips[["total_bill", "tip", "size", "smoker"]], hue="smoker")
plt.show()
```

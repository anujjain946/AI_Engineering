Slicing: allows us to extract the substring from a given string by specifying the range of indices.

Syntax:string[start:end:step]

By default, start = 0, end = n, step = 1

Positive Indexing

```python
string_1[0:11:2]
```

```python
string_1[::2]
```

```python
string_1[::]
```

Negative indexing: Capture the substring of a given string

```python
string_1[-1:-6:-1]
```

Reversal of a string

```python
string_1[::-1]
```

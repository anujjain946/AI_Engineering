Write a decorator that measures the time a function takes to excute

```python
import time

def time_tracker(func):
    def inner():
        start = time.time()
        func()
        end = time.time()
        print("Time taken:", end - start)
    return inner

@time_tracker
def load_data():
    time.sleep(2)
    print("Data loaded")

load_data()
```

Log function calls

Without decorator: You are repeating yourself and scaling this will hurt.

```python
def process_payment():
    print("Function started")
    print("Processing payment...")
    print("Function ended")

def generate_invoice():
    print("Function started")
    print("Generating invoice...")
    print("Function ended")
```

With Decorator

```python
def log_wrapper(func):
    def inner():
        print("Function started")
        func()
        print("Function ended")
    return inner


@log_wrapper
def generate_invoice():
    print("Generating invoice...")

generate_invoice()
```

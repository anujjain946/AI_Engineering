Armstrong Number

```python
# function definition
def is_armstrong(num):
  num_str = str(num)
  power = len(num_str)

  total = 0
  for digit in num_str:
    total = total + int(digit) ** power

  return total == num

# function calling
num = int(input("Enter your number: "))
if is_armstrong(num):
  print(f"{num} is an armstrong number")
else:
  print(f"{num} is not an armstrong number")
```

Decorators

Write a decorator that logs all the function calls

Without decorator: For every function we are repeating the same logic to log all the function calls and even we scale our functions, this will hurt

```python
def process_payment():
  print("Function started")
  print("Processing payment....")
  print("Function ended")


def generate_invoice():
  print("Function started")
  print("Generating invoice....")
  print("Function ended")
```

```python
def log_wrapper(func):
  def inner():
    print("Function started")
    func()
    print("Function ended")
  return inner

@log_wrapper
def generate_invoice():
  print("Generating invoice")

@log_wrapper
def process_payment():
  print("Payment processing")
```

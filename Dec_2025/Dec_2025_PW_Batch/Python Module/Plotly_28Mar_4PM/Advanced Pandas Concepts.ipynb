Basic Data Cleaning

1. Drop Unnamed:0 and id specific column in the above files.
2. Check the dtype of each column and if there is any inconsistency, handle that [specifically handle cost and paid column and convert them to float type].

```python
purch_data.dtypes
```

```python
purch_data = purch_data.drop("Unnamed: 0", axis=1)
prod_data = prod_data.drop("Unnamed: 0", axis=1)

prod_data['cost']=prod_data['cost'].str.replace(r'[\$,]','',regex=True)
prod_data['cost']=prod_data['cost'].astype(float)

purch_data["paid"] = purch_data["paid"].replace(r"[\$,]", "", regex=True).astype(float)
purch_data["amount"] = purch_data["amount"].astype(float)

purch_data["purch_date"]=pd.to_datetime(purch_data["purch_date"])
```

Evaluate the average cost of products per company

Summarization

1. Pivot Tables: It is a powerful tool that allows us to summarize and aggregate data in a flexible and customizable way.

2. CrossTab: In Pandas, it is used to compute a simple cross tabulation of two factors. It is similar to pivot tables but focuses on counting the frequencies.

3. Melting

1. index: array-like, Series, or list of arrays/Series
Values to group by in the rows.
Multiple index values can also be provided for multi-level indexing index=['Region', 'Product']
2. columns: array-like, Series, or list of arrays/Series
Values to group by in the columns.
3. values: array-like, optional
Array of values to aggregate.
4. aggfunc: function, optional, default numpy.mean
Multiple aggregation function can be used;  aggfunc=[sum, 'mean'].


```python
# sample data
data = {
    'region': ['east', 'west', 'east', 'west', 'east', 'west', 'east'],
    'product': ['A', 'A', 'B', 'B', 'C', 'C', 'B'],
    'sales': [100, 150, 200, 250, 300, 350, 100]
    }

data = pd.DataFrame(data)
# creating a pivot table
pivot_table = data.pivot_table(values = 'sales', index = 'region', columns = 'product', aggfunc='sum')
print(pivot_table)
```

```python
# sample data
data = {
    'region': ['east', 'west', 'east', 'west', 'east', 'west', 'east'],
    'product': ['A', 'A', 'B', 'B', 'C', 'C', 'B'],
    'sales': [100, 150, 200, 250, 300, 350, 100]
    }

data = pd.DataFrame(data)
# creating a crosstab
crosstab = pd.crosstab(data['region'], data['product'])
print(crosstab)
```

```python
data = {
    'region': ['east', 'west'],
    'prod_A': [100, 150],
    'prod_B': [200, 250],
    'prod_C': [300, 350]
}

df = pd.DataFrame(data)
melted_df = pd.melt(df, id_vars=['region'], value_vars=['prod_A', 'prod_B', 'prod_C'], var_name='Product', value_name='Sales')
print(melted_df)
```

Sorting

```python
arr1 = np.array([3, 1, 5, 2, 4, 9, 7, 8])
sorted_arr = np.sort(arr1)
print(sorted_arr)
```

```python
arr2 = np.array([[5, 2, 3],
                 [8, 1, 6],
                 [7, 4, 9]])

# sort the data along axis = 0 meaning column wise
sorted_arr_axis0 = np.sort(arr2, axis=0)
print(sorted_arr_axis0)
```

```python
# sort the data along axis = 1 meaning row wise
sorted_arr_axis1 = np.sort(arr2, axis=1)
print(sorted_arr_axis1)
```

```python
sorted_arr_default_axis = np.sort(arr2)
print(sorted_arr_default_axis)
```

Statistical Operations

```python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])

mean_all = np.mean(arr)
```

```python
print(mean_all)
```

```python
mean_column_wise = np.mean(arr, axis=0)
print(mean_column_wise)
```

```python
mean_row_wise = np.mean(arr, axis=1)
print(mean_row_wise)
```

```python
median_all = np.median(arr)
print(median_all)
```

```python
standard_deviation_all = np.std(arr)
print(standard_deviation_all)
```

```python
min_all = np.min(arr)
print(min_all)
```

```python
max_all = np.max(arr)
print(max_all)
```

```python
sum_all = np.sum(arr)
print(sum_all)
```

Simple Assignment Operator

```python
list1 = [1, 2, 3]
list2 = list1

print(list1)
print(list2)
```

In Python, id() is a built-in function that returns the unique identity of an object as an integer.

```python
print(id(list1))
print(id(list2))
```

Nested Lists

```python
list3 = [[1, 2], [3, 4]]

list3[0]
```

```python
list3[0][1]
```

```python
list3[1][1]
```

```python
list4 = list3
print(list3)
print(list4)
```

```python
list4.append([5, 6])

print(list3)
print(list4)
```

Copy in the lists

1. Shallow Copy
2. Deep Copy

```python
print(list3)
print(list4)
```

```python
list5 = list4.copy()

print(list4)
print(list5)
```

```python
print(id(list4))
print(id(list5))
```

Shallow Copy:

*   Outer list is new
*   Inner lists are shared



```python
list5[1][0] = 30

print(list4)
print(list5)
```

```python
list5.append([7,8])

print(list4)
print(list5)
```

```python
list5.remove([30, 4])
print(list4)
print(list5)
```

Deep Copy:


*   Outer list is new
*   Inner lists is also new



```python
import copy
list5
```

```python
list6 = copy.deepcopy(list5)
print(list5)
print(list6)
```

```python
list6[2][0] = 70

print(list5)
print(list6)
```

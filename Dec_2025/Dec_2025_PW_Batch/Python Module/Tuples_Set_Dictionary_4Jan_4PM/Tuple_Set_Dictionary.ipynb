Set: Unordered, mutable, unindexed and does not allow duplicate elements.

```python
set1 = {1, 2, 3, 2, 3, 2, 3, 2}
print(set1)
```

```python
mylist = [1, 2, 3, 4, 5, 2, 5, 3, 8]

myset = set(mylist)
print(myset)
```

```python
type(myset)
```

```python
myset.add(10)
print(myset)
```

```python
myset.remove(3)
print(myset)
```

Union Operation

```python
set1 = {1, 2, 3}
set2 = {4, 5, 6}

union_set = set1.union(set2)
print(union_set)
```

Intersection Operation

```python
set2
```

```python
set3 = {3, 4, 5}

intersection_set = set2.intersection(set3)
print(intersection_set)
```

Symmetric Difference: contains all the elements that are present in either set A or set B but not in both

```python
symmetric_difference_set = set2.symmetric_difference(set3)
print(symmetric_difference_set)
```

Difference: The difference between two sets A and B contains the elements that are present in set A but not in set B.

```python
set1
```

```python
set2
```

```python
set3
```

```python
diff_set = set1.difference(set3)
print(diff_set)
```

```python
diff_set = set3.difference(set1)
print(diff_set)
```

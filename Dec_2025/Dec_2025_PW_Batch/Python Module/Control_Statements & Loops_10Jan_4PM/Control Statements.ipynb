Nested if statements: using if statement insider another if and this concept is simply called as nesting.

Example 1: Club Entry Check

````python
age = int(input("Enter your age: "))
has_id = False # data type => boolean

if age >= 18:
  if has_id:
    print('You can enter the club.')
  else:
    print('You need an ID to enter.')
else:
  print('You are too young to enter')
````

Example 3: Password Strength Check

if the length of the password is more than or equal to 8 and it contains digits as well, in that case we consider the password as strong password otherwise if the length of the password is more than or equal to 8 but there is no digit in it, then we say it as weak password otherwise we can mention as weak and too short password.

password = "python123"

output = "Strong Password"

password = "pythonprogramming"

output = "Weak: Add a digit in the password"

password = "python"

output = "Weak: Too Short"

````python
password = input("Enter your password: ")
password_length = len(password)
has_digit = any(char.isdigit() for char in password)

if password_length >= 8 and has_digit:
  print("Strong password")
elif password_length >= 8 and not has_digit:
  print("Weak password: Add a digit in your password")
else:
  print("Weak and too short password")
````

Issue Of Infinite Loop

Problem: Since x > 0 is always true, this loop never stops. To avoid infinite loops:
* use proper conditions

* include increment/decrement logic

* use break statement whenever required

````python
x = 1

while x > 0:
  print('x is positive')
````

Control Transfer Statements: break, continue and pass

break: exit the code/loop immediately in a complete sense.

````python
i = 1

while i <= 10:
  if i == 5:
    break
  print(i)
  i = i + 1
````

````python
for i in range(11):
  if i == 7:
    break
  print(i)
````

continue statement: simply skips the current iteration

````python
for i in range(11):
  if i == 7:
    continue
  print(i)
````

````python
for num in range(1, 11):
  if num % 2 == 0:
    continue # skip them
  print(num) # remaining numbers will be displayed
````

pass statement: literally does nothing, it's simply a placeholder

````python
for i in range(1, 6):
  if i == 3:
    pass # placeholder to write some future logic
  print("values of i: ", i)
````

Nested Loops

````python
for i in range(1, 6):
  for j in range(1, 11):
    print(i * j)
````

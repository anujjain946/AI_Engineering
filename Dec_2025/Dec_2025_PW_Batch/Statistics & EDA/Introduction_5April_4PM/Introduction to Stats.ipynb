Import all the required libraries

```python
import numpy as np
import pandas as pd
```

Generate the fake data

nominal data

```python
np.random.seed(0)
num_samples = 1000
# create a nominal column
colors = ['red', 'blue', 'green', 'yellow', 'black']
nominal_data = np.random.choice(colors, num_samples)

# create dataframe
df = pd.DataFrame(nominal_data, columns=['Color'])
df
```

ordinal data

```python
satisfaction_levels = ['low', 'medium', 'high']
ordinal_data = np.random.choice(satisfaction_levels, num_samples, p = [0.3, 0.4, 0.3])
df['satisfaction_levels'] = ordinal_data
df
```

interval data

```python
np.random.seed(0)

# interval data => temperature in the range from 0 to 100
interval_data = np.random.randint(0, 100, num_samples)

df['temperature'] = interval_data
df
```

ratio data

```python
# generate the fake data
ratio_data = np.random.rand(num_samples) * 1000 # income in dollars

df['income'] = ratio_data
df
```

discrete data -> num_of_children

```python
discrete_data = np.random.randint(1, 10, num_samples)

df['number_of_children'] = discrete_data
df
```

continuous data -> weight

```python
continuous_data = np.random.rand(num_samples) * 10 # weights in kg
df['weights'] = pd.DataFrame(continuous_data, columns=['weights'])
df
```

```python
df.to_excel('types_of_data.xlsx')
```

Measures Of Dispersion

```python
data = pd.read_csv('/content/process_engineering_data.csv')
data.head()
```

Range Of temperature

```python
temp_range = data['temperature'].max() - data['temperature'].min()
print(temp_range)
```

Interquartile Range => Q3 - Q1

```python
Q1 = data['temperature'].quantile(0.25)
Q3 = data['temperature'].quantile(0.75)
temp_iqr = Q3 - Q1
print(temp_iqr)
```

Variance of the pressure column

```python
var_pressure = data['pressure'].var()
print(var_pressure)
```

Standard Deviation of the pressure column

```python
std_pressure = data['pressure'].std()
print(std_pressure)
```

Measures of Central Tendency

```python
data['pressure'].mean()
```

```python
data['pressure'].median()
```

```python
data['pressure'].mode()
```

Measure of skewness is done by skew() function.

```python
flow_rate_skewness = data['flow_rate'].skew()
```

Data Distribution => Right Skewed

```python
print(flow_rate_skewness)
```

```python
import random
from collections import Counter

# number of trials
trials = 10000
results = []


for i in range(trials):
  outcome = random.choice(['H', 'T'])

  if outcome == 'H':
    results.append(1) # win $1
  else:
    results.append(-1) # lose $1

# calculate the probabilities
counter = Counter(results)
probability_distribution = {outcome: count / trials for outcome, count in counter.items()}
print(f'Probability Distribution for coint toss: {probability_distribution}')
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)
heights = np.random.normal(loc=170, scale=10, size=1000) # mean = 170 cm, std = 10

plt.hist(heights, bins=15, edgecolor='black')
plt.title('Histogram of heights')
plt.xlabel('Height(cm)')
plt.ylabel('Frequency')
plt.show()
```

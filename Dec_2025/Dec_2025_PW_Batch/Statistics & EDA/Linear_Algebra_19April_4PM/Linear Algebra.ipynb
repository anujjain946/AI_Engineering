Matrix

1. Transpose Operation

A matrix representing the ratings of 3 users for 4 movies.

```python
ratings_matrix = np.array([
    [5, 4, 1, 2], # user 1 ratings
    [1, 2, 5, 4], # user 2 ratings
    [4, 3, 2, 5]  # user 3 ratings
])

print(ratings_matrix)
```

```python
transposed_matrix = ratings_matrix.T
print(transposed_matrix)
```

Cosine Similarity

```python
doc1 = np.array([30, 25, 20, 0, 1])
doc2 = np.array([5, 4, 3, 0, 0])
doc3 = np.array([0, 1, 0, 40, 30])

# calculate the cosine similarity of doc1 and doc2
dot_product_12 = np.dot(doc1, doc2)
norm_doc1 = np.linalg.norm(doc1)
norm_doc2 = np.linalg.norm(doc2)
similarity_12 = dot_product_12 / (norm_doc1 * norm_doc2)

# calculate the cosine similarity of doc2 and doc3
dot_product_23 = np.dot(doc2, doc3)
norm_doc2 = np.linalg.norm(doc2)
norm_doc3 = np.linalg.norm(doc3)
similarity_23 = dot_product_23 / (norm_doc2 * norm_doc3)


# calculate the cosine similarity of doc2 and doc3
dot_product_13 = np.dot(doc1, doc3)
norm_doc1 = np.linalg.norm(doc1)
norm_doc3 = np.linalg.norm(doc3)
similarity_13 = dot_product_13 / (norm_doc1 * norm_doc3)

print(f'Similarity(Doc1, Doc2): {similarity_12}')
print(f'Similarity(Doc2, Doc3): {similarity_23}')
print(f'Similarity(Doc1, Doc3): {similarity_13}')
```

```python
mat = np.array([[2.5, -0.5],
                [-0.5, 2.5]])

eigen_values, eigen_vectors = np.linalg.eig(mat)

print(f'Eigen Values are: {eigen_values}')
print(f'Eigen Vectors are: {eigen_vectors}')
```

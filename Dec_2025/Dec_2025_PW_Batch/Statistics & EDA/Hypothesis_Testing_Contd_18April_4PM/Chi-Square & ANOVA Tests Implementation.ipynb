```python
import numpy as np
from scipy import stats

# observed values from the table
observed_values = np.array([[50, 30, 20],
                            [30, 40, 30],
                            [20, 30, 50]])

# perform chi-square test of independence
chi2_stat, p_val, dof, expected_values = stats.chi2_contingency(observed_values)
```

```python
print(chi2_stat)
```

```python
print(dof)
```

```python
print(expected_values)
```

```python
print(p_val)
```

```python
alpha = 0.05

if p_val < alpha:
  print('Reject the H0 -> There is a significant association between age group and product preferences.')
else:
  print('Fail to reject the H0 -> There is a no association between age group and product preferences.')
```

ANOVA Test

```python
group_A = [105,110,98,107,103]
group_B = [130,125,132,128,129]
group_C = [120,115,117,119,121]


# perform one-way ANOVA
f_stats, p_val = stats.f_oneway(group_A, group_B, group_C)
```

```python
print(f_stats)
```

```python
print(p_val)
```

```python
alpha = 0.05

if p_val < alpha:
  print('Reject the H0')
else:
  print('Fail to reject the H0')
```

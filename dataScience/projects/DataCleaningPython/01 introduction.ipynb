```python
import json
```

```python
#lets write function for load the data
def loadData(filename):
    with open(filename,"r") as file:
        data = json.load(file)
    return data
    
    
```

```python
data = loadData("data.json")
data
```

```python
#Write a function to display users and their connection
def displayUsers():
    print("\n user's and there connection \n")
    for user in data['users']:
        print(f"use ID {user['id']} ,Name {user['name']} is friend with :{user['friends']} and like pages are {user['liked_pages']}")
    print("\n  page Information \n")
    for pages in data['pages']:
        print(f"page Id {pages['id']} with page {pages['name']}")
    
        
displayUsers() 
```

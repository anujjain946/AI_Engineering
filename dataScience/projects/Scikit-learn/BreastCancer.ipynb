```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import numpy as np
import pandas as pd
```

```python
#data loading
breast_cancer = load_breast_cancer()

target = pd.DataFrame(breast_cancer.data, columns=breast_cancer.feature_names)



label = pd.DataFrame(breast_cancer.target_names[breast_cancer.target], columns=['target'])


X_train, X_test, y_train, y_test = train_test_split(target, label, test_size=0.2, random_state=42)

model = RandomForestClassifier()
model.fit(X_train, y_train.values.ravel())
y_pred = model.predict(X_test)
print("Predicted labels:", y_pred)
print("Actual labels:", y_test.values.ravel())


accuracy_score = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy_score:.2f}")   


classification_rep = classification_report(y_test, y_pred)
print("Classification Report:")
print(classification_rep)


confusion_matrix = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(confusion_matrix) 


```

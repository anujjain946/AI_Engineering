#List and List Methods
- A list in Python is an ordered, mutable collection of elements. It can contain elements of different types.
- 

```python
#empy list
emp_list = []

#list with element
element_list = [1,2,3,4,5]

#list with datatype
data_type_list  = [1,2.33,'list',[1,2,3],True]
print(data_type_list)
```

```python
#Common List Methods
list = [1,2.33,'list',[1,2,3],True]
#1.append() use for Adds an element x to the end of the list.
list.append(False)
print(list)
print('\n')

#2. extend(iterable)	Extends the list by appending all elements from an iterable.
list.extend([2,3,4,5])
print(list)
print('\n')


#3. insert(index, x)	Inserts x at the specified index.
list.insert(0,55)
print(list)
print('\n')

#4. remove(x)	Removes the first occurrence of x in the list.
list.remove(55)
print(list)
print('\n')

#5. pop([index])	Removes and returns the element at index (last element if index is not provided).
list.pop(0)
print(list)
print('\n')

#6. index(x)	Returns the index of the first occurrence of x.
print(list.index(2))
print('\n')

#7. count(x)	Returns the number of times x appears in the list.
print(list.count(2))
print('\n')

#8. sort()	Sorts the list in ascending order.
l = [44,55,66,7,4,3]
l.sort()
print(l)
print('\n')

#9. reverse()	Reverses the order of the list.
l.reverse()
print(l)

#10. copy()	Returns a shallow copy of the list.
l2 = list.copy()
print(l2)
print(list)


```

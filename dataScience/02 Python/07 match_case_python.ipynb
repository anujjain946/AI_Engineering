#Match Case in Python
- match-case is used to check multiple conditions like switch-case in other languages.
It makes code cleaner, readable, and structured.

    match variable:
        case value1:
            # code
        case value2:
            # code
        case _:
            # default case



```python
code = 5
match code:
    case 200:
        print("Getting response")
    case 400:
        print("Bad request successfully.")
    case 404:
        print("Not Found")
    case 500:
        print("Internal server error")
    case _:
        print("Invalid code")
print("Program end")
        


```

#String in Python
- A string is a sequence of characters enclosed in single (' '), double (" "), or triple quotes (''' ''').

```python
lang = "python"
lang
```

```python
print(len(lang))
```

```python
#lower() / upper()
print(lang.lower())
print(lang.upper())
```

```python
#capitalize() / title()
print(lang.capitalize())
print(lang.title())
```

```python
#strip() / lstrip() / rstrip()
str = "  python "
print(str.strip()) #remove white space both site
print(str.lstrip()) #remove white space left site
print(str.rstrip()) #remove white space right site
```

```python
#replace()
str = "php is a simple language"
print(str.replace("php","python"))
```

```python
#split()
print(str.split())#reurn a list
```

```python
#join
words = ["Python", "is", "easy"]
print("-".join(words))
```

```python
#find() / index()
print("Python".find("t"))   # 2
print("Python".index("t"))  # 2
```

```python
#count()
print(str.count('p'))
```

```python
#startswith() / endswith()
print(str.startswith('php'))
print(str.endswith('language'))
```

```python
#isalpha() / isdigit() / isalnum()
print("abc".isalpha())
print("123".isdigit())
print("abc123".isalnum())
```

```python
#String Formatting
name = "Anuj"
age = 25
print(f"My name is {name} and age is {age}")
```

```python
#Escape Characters
print("Hello\nWorld")
print("He said \"Hello\"")
```

```python
#Loop Through String
for ch in "Python":
    print(ch)
```

```python
#Check Substring
print("Py" in "Python")   # True
```

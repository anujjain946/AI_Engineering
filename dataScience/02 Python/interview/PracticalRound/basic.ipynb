Q4:***Count Frequency of Characters***

```python
str = 'Python'
frequency = {}

for i in str:
    if i in frequency:
        frequency[i] += 1
    else:
        frequency[i] = 1
print(frequency)
```

Q9:***Sort List Without sort()***

```python
arr = [5, 2, 8, 1, 9]

for i in range(len(arr)):
    for j in range(i + 1, len(arr)):
        if arr[i] > arr[j]:
            arr[i], arr[j] = arr[j], arr[i]
print(arr)
```

Q11:***Flatten Nested List***

```python
#sing append method
nested = [[1,2],[3,4],[5,6]]

flat = []

for sublist in nested:
    for item in sublist:
        flat.append(item)

print(flat)
```

```python
#using extend method
nested = [[1,2],[3,4],[5,6]]

flat = []

for sublist in nested:
    flat.extend(sublist)

print(flat)
```

Q12:***Find Missing Number***

```python
lst = [1, 2, 3, 5]
n=5
expected_sum = n * (n + 1) // 2
print(f"expected sum is {expected_sum}")
actual_sum = sum(lst)
print(f"actual sum is {actual_sum}")    
missing_num = expected_sum - actual_sum
print(f"missing number is {missing_num}")
```

Q13:***Group Anagrams***

```python
words = ["eat","tea","tan","ate","nat","bat"]

result = {}

for word in words:
    key = ''.join(sorted(word))
    print(key)
    if key not in result:
        result[key] = []

    result[key].append(word)

print(list(result.values()))
```

Q13:***Second Largest Number***

```python
arr = [10, 50, 20, 80, 70]
first = second = float('-inf')

for num in arr:
    if num > first:
        second = first
        first = num
    elif num > second and num != first:
        second = num
print(f"largest number is {first}")
print(f"second largest number is {second}")
```

Q16:***Find First Non-Repeated Character in Python***

```python
def nonRepeated_char(s):
    char_count = {}
    # First loop counts how many times each character appears.
    for char in s:
        char_count[char] = char_count.get(char, 0) + 1

# Second loop checks characters in original order.
# The first character with count 1 is returned.
    for char in s:
        if char_count[char] == 1:
            return char
    return None
```

```python
s = "swiss"
print(nonRepeated_char(s))
```

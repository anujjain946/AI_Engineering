Summary of opp concept
- Class         :A blueprint for creating objects.            Class Car:
- Object        :An instance of a class.                      car1 = Car()
- Encapsulation :Restrct direct access to data.               self.__balance
- Inheritence   :A class inherits from another class.         Class Dog(Animal)
- Polymorphism  :using the same method in diffret ways        def fly(self)
- Abstraction   :Hiding the Implimentation details.           @abstrationmethod
- Magic Methods :Special method like __str__().               def __len__(self)
- Class Method  :Work with class variable.                    @classmethod
- static Method :Independent of class and instance.           @staticmethod


```python
class Employee:  #class
    company = "HP" #it's class attributes
    def __init__(self,name,salary):
        self.name = name #it's instance attributes
        self.salary = salary#it's instance attributes
    
    def printDetails(self):
        print(f"The company of {self.name} is {self.company} with salary {self.salary}")
    #using static method
    @staticmethod
    def printStatic():
        print("its static method")
    #using class method
    @classmethod
    def printClassMethod(cls):
        print(f"it is class attributes : {cls.company}")

emp =Employee("Harry",70000) #object
emp.printDetails()
emp.printStatic
emp.printClassMethod()
```

#Input in Python
- input() is used to take user input from the keyboard.
- It always returns data as a string.

```python
#Basic Input
lang = input("Enter the language")
print(lang)
```

```python
#Input with Type Casting
php_fee = int(input("enter the php fee"))
print(php_fee)
python_fee = float(input("enter the python fee"))
print(python_fee)
```

```python
#Multiple Inputs in One 
a, b = input("Enter two numbers: ").split()
print(a, b)
```

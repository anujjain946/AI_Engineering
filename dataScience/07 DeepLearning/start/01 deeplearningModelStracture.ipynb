Model Design Thinking

| #  | Question                   | Example Answer                                   |
| -- | -------------------------- | ------------------------------------------------ |
| 1  | Problem type?              | Binary Classification                            |
| 2  | Input features kitne hain? | 3                                                |
| 3  | Output kya hai?            | Pass/Fail                                        |
| 4  | Preprocessing kya chahiye? | Scaling                                          |
| 5  | Input shape kya hoga?      | `(3,)`                                           |
| 6  | Hidden layers kitni?       | Start with 1–3 layers                            |
| 7  | Activation?                | ReLU (hidden), Sigmoid/Softmax/Linear (output)   |
| 8  | Loss function?             | Depends on problem type                          |
| 9  | Optimizer?                 | Adam (good default)                              |
| 10 | Evaluation metric?         | Accuracy (classification), MAE/RMSE (regression) |

Import all the required libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Create the scatterplot

```python
tips = sns.load_dataset("tips")
tips.head()
```

This plot can give you an immediate sense of whether linear regression is suitable for the given data

```python
sns.regplot(x='total_bill', y='tip', data=tips)
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.title('Scatter Plot with Regression Line')
plt.show()
```

यह प्लॉट जल्दी से यह समझने में मदद करता है कि Linear Regression इस डेटा पर सही काम करेगी या नहीं। यदि अधिकांश डेटा पॉइंट्स एक सीधी रेखा (Regression Line) के आसपास हों और स्पष्ट बढ़ता या घटता ट्रेंड दिखे, तो Linear Regression उपयुक्त होती है। इस ग्राफ में Total Bill बढ़ने पर Tip भी बढ़ रही है, इसलिए Linear Regression एक अच्छा विकल्प लगता है।

Implementation of Regularization Techniques


```python
from sklearn.datasets import fetch_california_housing
houing = fetch_california_housing()

```

```python
houing
```

Data Splitting into independent features and target variable

```python
X, y = fetch_california_housing(return_X_y=True)

# independent feature
X
```

```python
# target feature
y
```

```python
X.shape
```

```python
y.shape
```

Train-Test Split

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=44)   
```

```python
x_train
```

```python
y_train
```

```python
x_train.shape
```

```python
x_test.shape
```

EDA => Statistical Module => Task to perform at your end


Data Leakage

Normalization of data


```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled
```

```python
x_test_scaled
```

Model Training

Linear Regression

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train_scaled, y_train)
```

```python
model.coef_
```

```python
model.intercept_
```

Lasso Regression (L1 Regularization)

```python
from sklearn.linear_model import Lasso
model_lasso = Lasso()
```

```python
model_lasso.fit(x_train_scaled, y_train)
```

```python
model_lasso.coef_
```

Ridge Regression (L2 Regualarization)

```python
from sklearn.linear_model import Ridge
model_ridge = Ridge()
```

```python
model_ridge.fit(x_train_scaled, y_train)
```

Elastic Net Regression (Combination of L1 and L2)

```python
from sklearn.linear_model import ElasticNet
model_elastic_net = ElasticNet()
```

```python
model_elastic_net.fit(x_train_scaled, y_train)
```

Model Prediction

```python
y_pred = model.predict(x_test_scaled)
print(y_pred)
```

```python
y_pred_Lasso = model_lasso.predict(x_test_scaled)
print(y_pred_Lasso)
```

```python
y_pred_Ridge = model_ridge.predict(x_test_scaled)
print(y_pred_Ridge)
```

```python
y_pred_Elastic = model_elastic_net.predict(x_test_scaled)
print(y_pred_Elastic)
```

Model Evaluation

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
print(mean_squared_error(y_test, y_pred))
```

```python
mean_absolute_error(y_test, y_pred)
```

```python
r2_score(y_test, y_pred)
```

Testing Set Error

```python
# linear regression335047$0
print(mean_squared_error(y_test, y_pred))
```

```python
y_pred_train = model.predict(x_train_scaled)
```

Training Set Error

```python
print(mean_squared_error(y_train, y_pred_train))
```

There is no overfitting, so no need to further go for regularization techniques

Import all the required libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#sklearn libraries
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score, recall_score
```


Load the data

```python
X, y = load_breast_cancer(return_X_y=True)
X
```

```python
y
```

Data Splitting based on train and test set

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=44)
```

Model Training

```python
# SVC => kernel and C
from sklearn.svm import SVC, SVR

model = SVC(kernel='linear', C=0.1)
model.fit(X_train, y_train)
```

Model Prediction

```python
y_pred = model.predict(X_test)
y_pred
```


Model Evaluation

```python
accuracy_score(y_test, y_pred)
```

```python
recall_score(y_test, y_pred)
```

Visualization

```python
X_vis = X[:, :2]
svm_vis = SVC()
svm_vis.fit(X_vis, y)
```

```python
plt.scatter(X_vis[:, 0], X_vis[:, 1], c=y, s=20)
plt.scatter(svm_vis.support_vectors_[:,0], svm_vis.support_vectors_[:,1], s=200, label='support vectors')
plt.title("SVM Decision Boundary")
plt.show()
```

Hyperparameter Tuning Techniques

GridSearchCV

```python
model = SVC()
from sklearn.model_selection import GridSearchCV

param_grid = {
    'C': [0.1, 1, 10, 100],
    'kernel': ['linear', 'rbf']
}

grid_search = GridSearchCV(
    estimator = model,
    param_grid = param_grid
)
```

```python
grid_search.fit(X_train, y_train)
```

RandomizedSearchCV


```python
model = SVC()
from sklearn.model_selection import RandomizedSearchCV

param_distr = {
    'C': [0.1, 1, 10, 100],
    'kernel': ['linear', 'rbf']
}

random_search = RandomizedSearchCV(
    estimator = model,
    param_distributions = param_distr
)
```

```python
random_search.fit(X_train, y_train)
```

```python
random_search.best_params_
```

Practical Improvement: Use Pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('svm', SVC())
])

param_grid = {
    'svm__C': [0.1, 1, 10, 100],
    'svm__kernel': ['linear', 'rbf']
}

grid_search = GridSearchCV(
    estimator = pipeline,
    param_grid = param_grid,
    cv = 5
)

grid_search.fit(X_train, y_train)
```

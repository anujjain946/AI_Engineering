**Multivariate analysis means analyzing three or more variables together.**

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np


data =  pd.read_csv("../dataset/Employee.csv")
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

data["Experience"] = np.random.randint(
    1,
    20,
    size=len(data)
)

data["Salary"] = (
    30000
    + data["Experience"] * 6000
    + np.random.randint(
        -5000,
        5000,
        size=len(data)
    )
)

# Correlation
corr = data[
    ["Experience", "Salary"]
].corr()

print(corr)

# Heatmap
plt.figure(figsize=(6, 4))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Experience vs Salary Correlation")

plt.show()
```

***Univariate Analysis***

- Univariate analysis means analyzing one variable at a time.

import library & laod dataset

```python
import pandas as pd
import matplotlib.pyplot as plt


data =  pd.read_csv("../dataset/Employee.csv")
```

```python
data.head()
```

Univariate Analysis

```python
print(data["Age"].describe())
```

Visualization

```python
figure = plt.figure(figsize=(10, 5))
bins = [20, 30, 40, 50, 60, 70, 80,100]

plt.hist(data["Age"], bins=20,edgecolor='black')

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")

plt.grid(True)
plt.show()
```

```python
print("Skewness:", data["Age"].skew())
```

check null


```python
data.isnull().sum()
```

Analyze outliers

```python
plt.boxplot(data["Age"].dropna())

plt.ylabel("Age")
plt.title("Age Outlier Detection")

plt.show()
```

```python
Q1 = data["Age"].quantile(0.25)
Q3 = data["Age"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = data[
    (data["Age"] < lower) |
    (data["Age"] > upper)
]

print("Number of outliers:", len(outliers))
print(outliers)
```

```python
# 5. Standard Deviation
# Standard deviation tells you how much the values vary around the mean.
std = data["Age"].std()
mean = data["Age"].mean()

print("Standard Deviation:", std)
print("Mean:", mean)
```

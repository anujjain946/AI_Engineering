```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures


# load the dataset
df = pd.read_csv("../dataset/Employee.csv")

df['Experience'] = np.random.randint(1, 20, size=len(df))

df['Salary'] = (
    30000
    + df['Experience'] * 6000
    + np.random.randint(-5000, 5000, size=len(df))
)
```

```python
X = df[["Experience"]]

y = df["Salary"]

poly = PolynomialFeatures(
    degree=2,
    include_bias=False
)

X_poly = poly.fit_transform(X)

model = LinearRegression()

model.fit(X_poly, y)
```

```python
### Prediction


new_data = [[6]]

new_data_poly = poly.transform(
    new_data
)

prediction = model.predict(
    new_data_poly
)

print(f'''Predicted salary for {new_data[0][0]} years of experience: ${prediction[0]:.2f}''')
```

```python
import numpy as np
import matplotlib.pyplot as plt

X = df[["Experience"]]
y = df["Salary"]

poly = PolynomialFeatures(
    degree=2,
    include_bias=False
)

X_poly = poly.fit_transform(X)

model = LinearRegression()

model.fit(X_poly, y)

x_range = np.linspace(
    X.min(),
    X.max(),
    100
).reshape(-1, 1)

x_range_poly = poly.transform(
    x_range
)

y_pred = model.predict(
    x_range_poly
)

plt.scatter(X, y)

plt.plot(
    x_range,
    y_pred
)

plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Polynomial Regression")

plt.show()
```

**Simple Linear Regression uses one independent feature to predict one continuous target.**

Example:

```text
Experience → Salary
```

Equation:

[
y = b0 + b1x
]

### Implementation

````python
from sklearn.linear_model import LinearRegression
import pandas as pd
import numpy as np

df = pd.read_csv("../dataset/Employee.csv")

df['Experience'] = np.random.randint(1, 20, size=len(df))

df['Salary'] = (
    30000
    + df['Experience'] * 6000
    + np.random.randint(-5000, 5000, size=len(df))
)


X = df[["Experience"]]

y = df["Salary"]

model = LinearRegression()

model.fit(X, y)

### Prediction

prediction = model.predict(
    [[6]]
)

print(prediction)
````

# 7. Understanding Coefficients

````python
print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_)
````

The model becomes:

[
Salary = 30108.51 + 5989.34(Experience)
]

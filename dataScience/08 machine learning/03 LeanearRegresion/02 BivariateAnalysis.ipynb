***Bivariate Analysis***
- Bivariate analysis means analyzing two variables together.


***import library & laod dataset***

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np


data =  pd.read_csv("../dataset/Employee.csv")
```

```python
data.head()
```

```python
# Add column Salary and Experience
data['Experience'] = np.random.randint(1, 20, size=len(data))

data['Salary'] = (
    30000
    + data['Experience'] * 6000
    + np.random.randint(-5000, 5000, size=len(data))
)
data.head()
```

```python
figure = plt.figure(figsize=(10, 6))
plt.scatter(
    data["Experience"],
    data["Salary"]
)

plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Experience vs Salary")

plt.show()
```

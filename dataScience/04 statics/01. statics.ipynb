***Measures of Central Tendency***


```python
#Import Libraries
import pandas as pd
import numpy as np
```

```python
#statics is science of analys,manipulate,predict data.
data = {
    "Student": ["A", "B", "C", "D", "E","F"],
    "Marks": [70, 80, 90, 85, 70,50]
}

#create data fram
students = pd.DataFrame(data)
print(students)
```

***Descriptive Statistics***
- Mean, Max, Min, Count

```python
#Descriptive :use for manipulation ,calculation of sampling data.
#mean
print("Average Marks", students["Marks"].mean())
#meadian
print("Median Marks", students["Marks"].median())
#mode
print("Mode Marks", students["Marks"].mode())
#Max
print("Maximum Marks", students["Marks"].max())
#Min
print("Minimum Marks", students["Marks"].min())
#count
print("Total Students", students["Marks"].count())
#veriance
print("Veriance", students["Marks"].var())
#standard deviation
print("Standard Deviation", students["Marks"].std())
```

***Inferential Statistics***
- Sample Survey


```python
#Inferential Statistics : use for calculate the data for predictions/decisions .
sample_data = np.random.choice([1,2,3,4,5], size=100)
print(sample_data)
print("Survey Average Rating:", np.mean(sample_data))
```

***Type of Data***

- Qualitative Data (Categorical)
    - Nominal Data
    - Ordinal Data
- Quantitative Data
    - Discrete Data
    - Continuous Data

```python
#Qualitative Data ->Nominal Data
#Non numeric data
#using only for maning and leveling it is unorder data.
sample = 1000
data  = np.random.choice(["Bhind","Gwalior","Indore","Bhopal","ratlam"], size=sample)
data = pd.DataFrame(data,columns=['City'])
print(data)
```

```python
#Qualitative Data ->Ordinal Data
#it is order data it is categoried data
data['rating'] = np.random.choice(["Good","poor","Excellent"], size=sample)
print(data)
```

```python
#Qualitative Data ->Discrete Data
#Countable numbers. Whole numbers only.
data['population'] = np.random.randint(100000, 400000, sample)
print(data)
```

```python
#Qualitative Data -> Continuous Data
#using decimal data only.
data['Area'] = np.random.uniform(100000, 400000, sample)
print(data)
```

***Sampling***

- Selecting small data from big population.
   - Random Sampling.
   - Systematic Sampling.
   - Stratified Sampling.
   - Cluster Sampling.

```python
#random sampling
#Everyone has equal chance.

population = pd.DataFrame({
    "People": range(100000,5000000)
})

sample = population.sample(5)

print(sample)
```

```python
#symentic sample
#Choose every nth item.
systematic_sample = population.iloc[::10]
print(systematic_sample)
```

```python
#Stratified Sampling
#Divide into groups then sample.
population['gender'] = pd.DataFrame(np.random.choice(["Male","Female"], size=5000000))

print(population.groupby("gender").sample(1))
```

```python
schools = {
    "School_A": 50,
    "School_B": 60,
    "School_C": 70
}

print("Selected Cluster:", schools["School_B"])
```

```python
#Measures of Dispersion
arr = [10, 20, 30, 40, 50]
print("Range:", max(arr) - min(arr))
print("Variance:", np.var(arr))
print("Standard Deviation:", np.std(arr))
print("Q1:", np.percentile(arr, 25))
print("Q3:", np.percentile(arr, 75))
#iqr
print("IQR:", np.percentile(arr, 75) - np.percentile(arr, 25))
```

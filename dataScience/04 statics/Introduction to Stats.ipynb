Import all the required libraries

```python
import numpy as np
import pandas as pd
```

Generate the fake data

Qualitative Data/nominal data

```python
np.random.seed(0)
num_samples = 1000
# create a nominal column
colors = ['red', 'blue', 'green', 'yellow', 'black']
nominal_data = np.random.choice(colors, num_samples)

# create dataframe
df = pd.DataFrame(nominal_data, columns=['Color'])
df
```

ordinal data


```python
satisfaction_levels = ['low', 'medium', 'high']
ordinal_data = np.random.choice(satisfaction_levels, num_samples, p = [0.3, 0.4, 0.3])
df['satisfaction_levels'] = ordinal_data
df
```

Quantitative Data/Discrete Data

```python
items = np.arange(1000,2000)
df['items'] = items
df
```

Quantitative Data/Continuous Data

```python
cost = np.random.uniform(10000,200000,num_samples)
df['cost'] = cost
df['cost'] = df['cost'].round(2)
df
```

write a file 

```python
df.to_csv('types_of_data.csv')
```

range of items

```python
range = df['items'].max() - df['items'].min()
print(range)
```

Interquartile Range => Q3 - Q1

```python
Q1 = df['items'].quantile(0.25)
Q3 = df['items'].quantile(0.75)
temp_iqr = Q3 - Q1
print(temp_iqr)
```

Variance of the cosr column

```python
std_pressure = df['cost'].std()
print(std_pressure)
```

Measures of Central Tendency

```python
df['cost'].mean()
```

```python
df['cost'].median()
```

```python
df['cost'].mode()
```

Measure of skewness is done by skew() function.

```python
flow_rate_skewness = df['cost'].skew()
flow_rate_skewness
```

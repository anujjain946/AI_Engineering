***1. Probability & Random Variables***


    

- Probability tells the chance of an event happening
- A variable whose value depends on chance


***2.Key Functions***
- PMF (Probability Mass Function)
    -  Used for discrete data [whole value]
        - Example:
            Dice probability
            P(X=1) = ⅙

- PDF (Probability Density Function)
    -  Used for continuous data
        - Example:
            Height distribution

-  CDF (Cumulative Distribution Function)
    -  Probability up to a value
        - Example:
            P(X ≤ 50 marks)




***3.Discrete & Continuous Distributions***
-  Discrete Distributions (countable values)
    - Bernoulli Distribution
        - Only 2 outcomes: success/failure
            - Example:Pass/FailHead/Tail

    - Binomial Distribution
        - Multiple Bernoulli trials
            - Example:10 coin toss → how many heads?
        
    - Poisson Distribution
        - Count events in fixed time
            - Example:Number of calls per hour,Customers per day


-  Continuous Distributions
    - Uniform Distribution
        -  All values equally likely
            - Example:Random number between 1–10
    - Normal Distribution (Gaussian)
        -  Bell-shaped curve
            - Example:Mean = Median = Mode
                - Height of people,Exam marks

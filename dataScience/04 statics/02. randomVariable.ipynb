```python
import random
from collections import Counter

#Number of trials
trials = 10000
rerults = []


for i in range(trials):
    outcomes = random.choice(['H','T'])
    # print(outcomes)
    if outcomes == 'H':
        rerults.append(1)
    else:
        rerults.append(-1)
# print(rerults)
#Calcullate probability
outcome_counts = Counter(rerults)
print(f"outcome counts : {outcome_counts}")
probability_distribution = {outcome: count / trials for outcome, count in outcome_counts.items()}
print(f"probability of coin : {probability_distribution}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)
heights = np.random.normal(loc=170, scale=10, size=1000) # mean = 170 cm, std = 10

plt.hist(heights, bins=15, edgecolor='black')
plt.title('Histogram of heights')
plt.xlabel('Height(cm)')
plt.ylabel('Frequency')
plt.show()
```

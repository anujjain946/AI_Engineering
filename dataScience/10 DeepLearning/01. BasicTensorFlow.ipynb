01. Basic

```python
#check vertion
print(tf.__version__)

#Scaler =>0D
a = tf.constant(10)
print(a)

#Vector => 1D tensor
print(tf.constant([1,2,3]))

#Matrix = 2D tensor
matrix =  tf.constant([[1,2,3],[4,5,6],[7,8,9]])
print(matrix)

#shape
print(matrix.shape)

#Data Type
print(matrix.dtype)

#Note : Constant Can't change value but into variable we can do it.
#Variable
weight = tf.Variable(10)
print(weight)

#Value Change
weight.assign(20)
print(weight)


```

```python
#add
a = tf.constant(50)
b = tf.constant(60)

print(f'Addition of two{a},{b} numbers is {a+b}')

print(f'Subtraction of two{a},{b} numbers is {a-b}')

print(f'multiply of two{a},{b} numbers is {a*b}')

print(f'deverion of two{a},{b} numbers is {a/b}')
```

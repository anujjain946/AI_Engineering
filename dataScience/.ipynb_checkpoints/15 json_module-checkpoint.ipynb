JSON
- json:javaScript Object Notation is lightweight data format used for data exchange between servers and applications.it iswidely used in APIs,web application and configrations

```python
#Import the json module
import json
```

```python
obj = {"name":"Python","company":"energy7","salary":"45000"}
json.dumps(obj) #dump as a string
data = json.dumps(obj)
type(data)
# print(data)
```

```python
#using json.loads() method 
#it's use for conveert str to disc
disc = json.loads(data)
type(disc)
# print(disc)
```

```python
#use the dump for data dump into the file
with open("data.json","w") as file:
    data = json.dump(obj,file)
    
    
```

```python
#read json file data using load
with open("data.json", "r") as file:
    data = json.load(file)
print(data)
```

```python
#Pretty JSON (Readable format 😍)
#write into the file 
    #json.dump(data, file, indent=4)

# data = json.dumps(data, indent=4)
# data = json.dumps(data, default=str)
data =json.dumps(data, sort_keys=True)
print(data)
```

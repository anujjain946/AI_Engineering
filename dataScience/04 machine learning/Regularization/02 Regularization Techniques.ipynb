Data Leakage


Normalization of data

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_tain)
X_test_scaled = scaler.transform(X_test)
```

```python
print(X_train_scaled)
```

```python
X_test_scaled
```

Model Training

Linear Regression

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train_scaled, y_train)
```

```python
model.coef_
```

```python
model.intercept_
```

Lasso Regression (L1 Regularization)

```python
from sklearn.linear_model import Lasso

model_lasso = Lasso()
model_lasso.fit(X_train_scaled, y_train)
```

```python
model_lasso.coef_
```

Ridge Regression (L2 Regualarization)

```python
from sklearn.linear_model import Ridge
model_ridge = Ridge()
```

```python
model_ridge.fit(X_train_scaled, y_train)
```

Elastic Net Regression (Combination of L1 and L2)

```python
from sklearn.linear_model import ElasticNet
model_elastic_net = ElasticNet()
```

```python
model_elastic_net.fit(X_train_scaled, y_train)
```

Model Prediction

```python
y_pred = model.predict(X_test_scaled)
print(y_pred)
```

```python
y_pred_Lasso = model_lasso.predict(X_test_scaled)
print(y_pred_Lasso)
```

```python
y_pred_Ridge = model_ridge.predict(X_test_scaled)
print(y_pred_Ridge)
```

```python
y_pred_Elastic = model_elastic_net.predict(X_test_scaled)
print(y_pred_Elastic)
```

Model Evaluation

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
print(mean_squared_error(y_test, y_pred))
```

```python
mean_absolute_error(y_test, y_pred)
```

```python
mean_squared_error(y_test, y_pred_Lasso)
```

```python
r2_score(y_test, y_pred_Lasso)
```

Testing Set Error

```python
# linear regression
print(mean_squared_error(y_test, y_pred))
```

```python
y_pred_train = model.predict(X_train_scaled)
```

Training Set Error

```python
print(mean_squared_error(y_train, y_pred_train))
```

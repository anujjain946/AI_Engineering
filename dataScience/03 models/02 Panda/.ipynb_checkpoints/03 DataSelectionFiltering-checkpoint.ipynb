```python
#Import pandas
import pandas as pd
```

```python
df = pd.read_csv("data.csv")
df
```

```python
#Selecting Columns
print(df["Actor"])

```

```python
#Selecting Multiple Columns
print(df[["Actor", "Film"]])
```

```python
#Selecting Rows by Index

df.loc[0] #By the row
```

```python
df.iloc[0]   #By the position
```

```python
#Select Specific Rows and Columns
df.loc[0, "Film"]        # Value at row 0, column 'Name'
df.iloc[0, 1]            # Value at row 0, column at index 1
```

```python
df.loc[0:2,["Actor","Film"]] # Rows 0 to 2, selected columnsd
```

```python
df.iloc[0:2, 0:2]              # Rows and cols by index position
```

```python
#Fast Access: .at and .iat
df.at[0, "Actor"]       # Fast label-based access
df.iat[0, 1]           # Fast position-based access
```

```python
#Filtering with Conditions
df[df["IMDb"] > 6]['Actor']

```

```python
#Multiple Conditions (AND / OR)
df[(df["IMDb"] > 5) & (df["Actor"] == "Shah Rukh Khan")]
# df[(df["Name"] == "Bob") | (df["Age"] < 30)]
```

***Querying with .query()***
1. Column names become variables
2. String values must be in quotes
3. Use backticks for column names with spaces or special characters
4. You can use @ to reference Python variables
5. Logical operators
6. Chained comparisons
7. Avoid using reserved keywords as column names
8. Case-sensitive
9. .query() returns a copy, not a view

```python
df.query("IMDb > 5 and Actor == 'Akshay Kumar'")
```

```python
#Dynamic column names:
col = "Actor"
df.query(f"{col} == 'Akshay Kumar'")
```

```python
#1. Column names become variables
df.query("IMDb > 5 and Actor == 'Akshay Kumar'")
```

```python
#2. String values must be in quotes
df.query("Actor == 'Akshay Kumar'")
```

```python
#3Use backticks for column names with spaces or special characters
#df.query("`first name` == 'Alice'")
```

```python
#4. You can use @ to reference Python variables
IMDb = 6
df.query("IMDb > @IMDb")
```

```python
#5. Logical operators
#df.query("age > 30 and city == 'Delhi'")  # ✅
```

```python
#6. Chained comparisons
df.query("5 < IMDb <= 9")
```

```python
#7. Avoid using reserved keywords as column names
#If you have a column named class, lambda, etc., you’ll need to use backticks:
#df.query("`class` == 'Physics'")
```

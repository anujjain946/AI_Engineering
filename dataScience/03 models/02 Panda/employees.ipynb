Remove extra spaces from text columns (if any).

````python
for col in emp.select_dtypes(include=['object']).columns:
    emp[col] = emp[col].str.strip().str.replace(r'\s+', ' ', regex=True)
````

````python
emp['City'] =  emp['City'].str.upper()

emp.head()
````

Convert department names to lowercase.

````python
emp['Department'] = emp['Department'].str.lower()

emp['Department']
````

Create a Bonus column (10% of Salary).

````python
emp['Bonus'] = (emp['Salary'] * 0.10).round(2)
emp.head()
````

Create an AnnualSalary column.

````python
# Create an AnnualSalary column.
emp['AnnualSalary'] = round(emp['Salary'] * 12,2)
emp.head()
````

```
Create an ExperienceCategory column:
Junior (<3 years)
Mid (3–6 years)
Senior (>6 years)
```


````python
# 1. Initialize the column with a default value
emp['ExperienceCategory'] = 'Unknown'

# 2. Assign 'Junior' to rows where Experience is less than 3
emp.loc[emp['Experience'] < 3, 'ExperienceCategory'] = 'Junior'

# 3. Assign 'Mid' to rows where Experience is between 3 and 6 (inclusive)
emp.loc[(emp['Experience'] >= 3) & (emp['Experience'] <= 6), 'ExperienceCategory'] = 'Mid'

# 4. Assign 'Senior' to rows where Experience is greater than 6
emp.loc[emp['Experience'] > 6, 'ExperienceCategory'] = 'Senior'

# View the results
print(emp[['Name', 'Experience', 'ExperienceCategory']].head())
````

***Adding Two Lists vs. NumPy Arrays***




```python
import numpy as np
import time

# Python list
size = 1_000_000
list1 = list(range(size))
list2 = list(range(size))

start = time.time()
result = [x + y for x, y in zip(list1, list2)]
end = time.time()
print("Python list addition time:", end - start)

# NumPy array
arr1 = np.array(list1)
arr2 = np.array(list2)

start = time.time()
result = arr1 + arr2  # Vectorized operation
end = time.time()
print("NumPy array addition time:", end - start)
```

```python
import time
import threading
import sys

# input seconds
total_seconds = int(input("Enter time in seconds: "))

paused = False

def listen_input():
    global paused
    while True:
        cmd = input().lower()
        if cmd == "p":
            paused = True
        elif cmd == "r":
            paused = False

# run input listener in background
threading.Thread(target=listen_input, daemon=True).start()

while total_seconds >= 0:

    if not paused:
        mins = total_seconds // 60
        secs = total_seconds % 60
        timer = f"{mins:02d}:{secs:02d}"

        print(timer, end="\r")
        time.sleep(1)
        total_seconds -= 1
    else:
        time.sleep(0.1)

# beep sound
print("\nTime's up!")
print("\a")  # system beep
```

```python
#Impot Time model
import time
```

```python
#Current Time (timestamp)
#getting time in second epoch
print(time.time())
```

```python
#Sleep / Delay

print("start")
time.sleep(3)
print("End after 3 sec")
```

```python
#Local Time ***Readable format me current time:
print(time.localtime())
print(time.ctime())
```

```python
#Custom Format Time
print(time.strftime("%Y-%m-%d %H:%M:%S"))
```

```python
#Execution Time Measure

start = time.time()

for i in range(10000):
    pass
end =time.time()
print("Execution time:", end - start)
```

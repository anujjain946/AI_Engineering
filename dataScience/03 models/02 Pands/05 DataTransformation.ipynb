***Sorting & Ranking***

```python
import pandas as pd
```

```python
data = pd.read_csv("data_cleaning_sample.csv")
data
```

```python
data.sort_values("Age")
```

```python
data.sort_values("Age", ascending=False) # Descending
```

```python
data.sort_values(["Age", "Gender"]) # Sort by multiple columns
```

***Reset Index***


```python
data.reset_index(drop=True, inplace=True)
# Reset the index and drop the old index
```

```python
data
```

***Sort by Index***

```python
data.sort_index()
```

```python
data["Rank"] = data["Age"].rank()
data
```

```python
data["Rank"] = data["Age"].rank(method="dense") # 1, 2, 2, 3
data
```

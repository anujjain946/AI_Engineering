2. load dataset

```python
data = pd.read_csv('batter.csv')  # Load your data into a DataFrame
print(data.head())  # Display the first few rows of the DataFrame
```

2. Basic Line Plot

```python

x = data[data['runs'] > 5000] 



# Get the 'batter' column for those specific rows
level = x['batter'] 
value = x['runs']


print(value)

# plt.plot(value, level,'o:r')
plt.plot(value, level, color='m', marker='o', linestyle='dashed', linewidth=0.5, markersize=12)
plt.show()
```

3. Customize Line Plot

```python
x = data[data['runs'] > 5000]
# Get the 'batter' column for those specific rows
level = x['batter'] 
value = x['runs']

plt.plot(value, level, color='green', marker='o',markersize=5, linestyle='-', linewidth=1)
plt.title('Batter Runs')
plt.xlabel('Runs')
plt.ylabel('Batter')
plt.legend()
plt.grid()

plt.show()
```

4. Multiple Lines

```python

x = [1, 2, 3, 4, 5]

sales_2024 = [10, 20, 15, 30, 40]
sales_2025 = [15, 25, 20, 35, 50]

plt.plot(x, sales_2024, label="2024",color='blue', marker='o', markersize=5, linestyle='--', linewidth=1)
plt.plot(x, sales_2025, label="2025",color='red', marker='o', markersize=5, linestyle='--', linewidth=1)

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Sales Comparison")

plt.legend()
plt.grid()
plt.show()
```

5. Bar Chart
- Used for categorical data.

```python
filtered_data = data[data['runs'] > 5000]

plt.figure(figsize=(10, 6))
# 1. Create a list of colors: red if runs < 6000, else skyblue
colors = ['red' if run < 6000 else 'skyblue' for run in filtered_data['runs']]

plt.bar(
        x=filtered_data['batter'],       # X-axis categories
        height=filtered_data['runs'],    # Y-axis data values
        width=0.6,                       # Custom bar width (slightly skinnier)
        bottom=0,                        # Base level of bars
        align='center',                  # Align bars on the center of the tick
        color=colors,                 # Fill color (or list of colors)
        edgecolor='black',               # Border color
        linewidth=1.5,                   # Border thickness
        alpha=0.5,                       # Transparency scale (0 to 1)
        yerr=100,                        # Example error bar margin
        capsize=5                        # Error bar cap size
)

plt.title('Batter Runs Greater than 5000')
plt.xlabel('Batter')
plt.ylabel('Runs')
plt.xticks(rotation=45)
plt.tight_layout()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

print(filtered_data)
```

6.Horizontal bar:


```python
# Your original data filtering
filtered_data = data[data['runs'] > 5000]

plt.figure(figsize=(10, 6))

# Dynamic color logic (Red if < 6000, green if >= 6000)
colors = ['red' if run < 6000 else 'green' for run in filtered_data['runs']]

# FIX: Changed yerr to xerr for horizontal alignment
plt.barh(
    filtered_data['batter'], 
    filtered_data['runs'], 
    color=colors, 
    edgecolor='black', 
    linewidth=1.5, 
    alpha=0.7,          # Bumped up to 0.7 so the colors are a bit richer
    xerr=100,           # Changed from yerr to xerr
    capsize=5
)

# Axis titles and chart titles
plt.title('Batter Runs Greater than 5000')
plt.ylabel('Batter')
plt.xlabel('Runs')

# FIX: Changed axis='y' to axis='x' to match horizontal bar lengths
plt.grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()

print(filtered_data)
```

7. Multiple Bar Charts.

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(10, 6))

products = ["Laptop", "Mobile", "Tablet"]

sales_2024 = [50, 80, 40]
sales_2025 = [60, 90, 55]
sales_2026 = [70, 100, 60]

x = np.arange(len(products))
width = 0.25  # Increased width slightly to look better filled

# FIX: Spread the bars evenly around the central tick mark 'x'
plt.bar(x - width, sales_2024, width, label="2024", edgecolor='black')
plt.bar(x,         sales_2025, width, label="2025", edgecolor='black')
plt.bar(x + width, sales_2026, width, label="2026", edgecolor='black')

plt.xticks(x, products)

plt.title('Calculate Sales')
plt.ylabel('Number of Units Sold')
plt.xlabel('Products')

# Optional: Add a lighter background grid for better readability
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.legend()

plt.show()
```

7. Pie Chart
- Used to show percentage/proportion.

```python
plt.figure(figsize=(8, 10))
labels = ["Python", "Java", "PHP", "C++"]
values = [40, 30, 20, 10]
myexplode = [0.1, 0, 0, 0]

plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    startangle = 90,
    explode=myexplode,
    shadow = True

)

plt.title("Programming Language Usage")
plt.legend(title = "Languages:")
plt.show()
```

8. Histogram
- Used to understand distribution of numerical data.

```python
filtered_data = data['runs'] 

plt.figure(figsize=(10, 6))

plt.hist(filtered_data, bins=5, color='skyblue', edgecolor='black')

plt.title("Runs Distribution")
plt.xlabel("Runs")
plt.ylabel("Frequency")

plt.show()
```
